## 1. Function Definition

A function is a reusable block of code designed to perform a specific task.

### def Keyword

In Python, functions are created using the `def` keyword.

Example:

def greet():
    print("Hello!")


The function can then be called using:

greet()

### Parameters

Parameters are variables written inside the function definition. They allow a function to receive input values.

Example:

def greet(name):
    print("Hello", name)

Here, `name` is a parameter.

### Return Values

The `return` statement sends a value back from a function to the code that called it.

Example:

def add(a, b):
    return a + b

result = add(5, 3)

The value returned is 8.

### None Return

If a function does not have a return statement, Python automatically returns `None`.

Example:

def greet():
    print("Hello!")

result = greet()

The value of `result` is None.

## 2. Arguments

Arguments are the actual values passed to a function when the function is called.

### Positional Arguments

Positional arguments are matched with parameters according to their position.

Example:

def introduce(name, age):
    print(name, age)

introduce("Areeba", 22)

Here, "Areeba" is assigned to `name` and 22 is assigned to `age`.

### Keyword Arguments

Keyword arguments specify the parameter name while calling the function.

Example:

introduce(name="Areeba", age=22)

This makes the function call more readable.

### Default Parameters

A default parameter has a predefined value that is used when no argument is provided.

Example:

def greet(name="User"):
    print("Hello", name)

greet()

The output will be:

Hello User

### Argument Unpacking

Argument unpacking allows values from a list or tuple to be passed to a function using `*`.

Example:

def add(a, b, c):
    return a + b + c

numbers = [10, 20, 30]

result = add(*numbers)

The `*` operator unpacks the list into separate arguments.

Python also supports `**` for unpacking dictionaries into keyword arguments.

## 3. Scope

Scope determines where a variable can be accessed in a Python program.

### Local Variables

A variable created inside a function is called a local variable. It can normally be accessed only inside that function.

Example:

def calculate():
    result = 100
    print(result)

Here, `result` is a local variable.

### Global Variables

A variable created outside a function is called a global variable. It can be accessed from different parts of the program.

Example:

name = "Areeba"

def greet():
    print(name)

### global Keyword

The `global` keyword allows a function to modify a global variable.

Example:

counter = 0

def increase():
    global counter
    counter += 1

### Variable Lifetime

Variable lifetime means how long a variable exists in memory.

A local variable normally exists while its function is executing. After the function finishes, the local variable is no longer accessible from outside the function.

A global variable generally exists for the lifetime of the program.

## 4. Function Design Principles

Good functions should be simple, readable, reusable, and focused on one task.

### Single Responsibility

A function should perform one specific task instead of trying to perform many unrelated tasks.

Example:

def calculate_total(price, quantity):
    return price * quantity

This function has one clear responsibility: calculating the total price.

### Descriptive Names

Function names should clearly describe what the function does.

Good example:

def calculate_average(numbers):
    ...

Poor example:

def do_task(numbers):
    ...

Descriptive names make code easier to understand.

### Docstrings

A docstring explains the purpose of a function. It is written inside the function using triple quotes.

Example:

def calculate_square(number):
    """Return the square of a number."""
    return number ** 2

Docstrings make functions easier to understand and maintain.

### Reusability

A good function should be reusable instead of repeating the same code multiple times.

Functions help reduce code duplication and improve program organization.

import math


def validate_numbers(numbers):
    """Validate that numbers is a non-empty list of numeric values.

    Parameters:
        numbers (list): List containing numeric values.

    Returns:
        None: Returns None if validation is successful.

    Raises:
        ValueError: If the list is empty or contains non-numeric values.
        TypeError: If the input is not a list.
    """
    if not isinstance(numbers, list):
        raise TypeError("Input must be a list.")

    if not numbers:
        raise ValueError("The list cannot be empty.")

    if not all(isinstance(num, (int, float)) and not isinstance(num, bool)
               for num in numbers):
        raise ValueError("All values must be numeric.")

```python
def calculate_average(numbers, precision=2):
    """Calculate the average of a list of numbers.

    Parameters:
        numbers (list): List of numeric values.
        precision (int): Number of decimal places for the result.

    Returns:
        float: The calculated average.

    Example:
        calculate_average([10, 20, 30])
        20.0
    """
    validate_numbers(numbers)

    if not isinstance(precision, int) or precision < 0:
        raise ValueError("Precision must be a non-negative integer.")

    average = sum(numbers) / len(numbers)
    return round(average, precision)


def find_median(numbers, precision=2):
    """Calculate the median of a list of numbers.

    Parameters:
        numbers (list): List of numeric values.
        precision (int): Number of decimal places for the result.

    Returns:
        float: The median value.

    Example:
        find_median([10, 20, 30])
        20.0
    """
    validate_numbers(numbers)

    if not isinstance(precision, int) or precision < 0:
        raise ValueError("Precision must be a non-negative integer.")

    sorted_numbers = sorted(numbers)
    length = len(sorted_numbers)
    middle = length // 2

    if length % 2 == 0:
        median = (
            sorted_numbers[middle - 1] + sorted_numbers[middle]
        ) / 2
    else:
        median = sorted_numbers[middle]

    return round(median, precision)


def get_standard_deviation(numbers, precision=2):
    """Calculate the population standard deviation.

    Parameters:
        numbers (list): List of numeric values.
        precision (int): Number of decimal places for the result.

    Returns:
        float: The population standard deviation.

    Example:
        get_standard_deviation([10, 20, 30])
    """
    validate_numbers(numbers)

    if not isinstance(precision, int) or precision < 0:
        raise ValueError("Precision must be a non-negative integer.")

    average = sum(numbers) / len(numbers)
    variance = sum((num - average) ** 2 for num in numbers) / len(numbers)

    return round(math.sqrt(variance), precision)
```

## Task 2: Text Processing Functions

In this task, functions are created for common text-processing operations.

Functions:
- count_words()
- extract_emails()
- remove_punctuation()
- title_case()
- process_text()

The `process_text()` function chains multiple functions together.

The task also demonstrates returning multiple values using a tuple.

```python
import re
import string


def count_words(text):
    """Count the number of words in a text.

    Parameters:
        text (str): Input text.

    Returns:
        int: Number of words.
    """
    if not isinstance(text, str):
        raise TypeError("Text must be a string.")

    return len(text.split())


def extract_emails(text):
    """Extract email addresses from text.

    Parameters:
        text (str): Input text.

    Returns:
        list: List of email addresses found in the text.
    """
    if not isinstance(text, str):
        raise TypeError("Text must be a string.")

    pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"
    return re.findall(pattern, text)


def remove_punctuation(text):
    """Remove punctuation from text.

    Parameters:
        text (str): Input text.

    Returns:
        str: Text without punctuation.
    """
    if not isinstance(text, str):
        raise TypeError("Text must be a string.")

    return text.translate(str.maketrans("", "", string.punctuation))


def title_case(text):
    """Convert text to title case.

    Parameters:
        text (str): Input text.

    Returns:
        str: Text converted to title case.
    """
    if not isinstance(text, str):
        raise TypeError("Text must be a string.")

    return text.title()
```

```python
def process_text(text, remove_punct=True, to_title=True):
    """Process text using optional punctuation removal and title case.

    Parameters:
        text (str): Input text.
        remove_punct (bool): Remove punctuation if True.
        to_title (bool): Convert text to title case if True.

    Returns:
        tuple: Word count, character count, and unique word count.
    """
    if not isinstance(text, str):
        raise TypeError("Text must be a string.")

    processed_text = text

    if remove_punct:
        processed_text = remove_punctuation(processed_text)

    if to_title:
        processed_text = title_case(processed_text)

    words = processed_text.split()
    word_count = len(words)
    char_count = len(processed_text)
    unique_words = len(set(word.lower() for word in words))

    return word_count, char_count, unique_words
```

```python
text = (
    "Hello, my name is Areeba! "
    "Contact me at areeba@example.com. "
    "I am learning Python."
)

print("Word Count:", count_words(text))
print("Emails:", extract_emails(text))
print("Without Punctuation:", remove_punctuation(text))
print("Title Case:", title_case(text))

result = process_text(text)

print("\nProcessed Text Statistics:")
print("Word Count:", result[0])
print("Character Count:", result[1])
print("Unique Words:", result[2])
```

## Task 3: Validation Function Suite

This task creates separate validation functions for:
- Email
- Phone number
- Date
- Password

Each validator returns a tuple:

(is_valid, error_message)

A wrapper function called `validate_user_input()` is used to call the appropriate validator.

```python
import re
from datetime import datetime


def validate_email(email):
    """Validate an email address.

    Returns:
        tuple: (is_valid, error_message)
    """
    if not isinstance(email, str) or not email.strip():
        return False, "Email cannot be empty."

    pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

    if not re.match(pattern, email):
        return False, "Invalid email format."

    return True, ""


def validate_phone(phone):
    """Validate a phone number.

    Returns:
        tuple: (is_valid, error_message)
    """
    if not isinstance(phone, str) or not phone.strip():
        return False, "Phone number cannot be empty."

    pattern = r"^\+?[0-9]{10,15}$"

    if not re.match(pattern, phone):
        return False, "Phone must contain 10 to 15 digits."

    return True, ""


def validate_date(date_string):
    """Validate a date in YYYY-MM-DD format.

    Returns:
        tuple: (is_valid, error_message)
    """
    if not isinstance(date_string, str) or not date_string.strip():
        return False, "Date cannot be empty."

    try:
        datetime.strptime(date_string, "%Y-%m-%d")
        return True, ""
    except ValueError:
        return False, "Date must be in YYYY-MM-DD format."


def validate_password(password):
    """Validate password strength.

    Requirements:
        - At least 8 characters
        - One uppercase letter
        - One lowercase letter
        - One digit

    Returns:
        tuple: (is_valid, error_message)
    """
    if not isinstance(password, str):
        return False, "Password must be a string."

    if len(password) < 8:
        return False, "Password must contain at least 8 characters."

    if not re.search(r"[A-Z]", password):
        return False, "Password must contain an uppercase letter."

    if not re.search(r"[a-z]", password):
        return False, "Password must contain a lowercase letter."

    if not re.search(r"[0-9]", password):
        return False, "Password must contain a digit."

    return True, ""
```

```python
def validate_user_input(input_type, value):
    """Call the appropriate validator based on input type.

    Parameters:
        input_type (str): Type of input to validate.
        value: Value that needs validation.

    Returns:
        tuple: (is_valid, error_message)
    """
    validators = {
        "email": validate_email,
        "phone": validate_phone,
        "date": validate_date,
        "password": validate_password,
    }

    if input_type not in validators:
        return False, "Unknown input type."

    return validators[input_type](value)
```

```python
test_inputs = [
    ("email", "areeba@example.com"),
    ("email", "invalid-email"),
    ("phone", "+923001234567"),
    ("phone", "123"),
    ("date", "2026-08-19"),
    ("date", "2026-15-40"),
    ("password", "Python123"),
    ("password", "weak"),
]

for input_type, value in test_inputs:
    is_valid, error_message = validate_user_input(
        input_type,
        value
    )

    print(f"{input_type}: {value}")
    print("Valid:", is_valid)
    print("Message:", error_message)
    print("-" * 40)
```
