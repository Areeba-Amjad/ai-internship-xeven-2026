## 1. Dunder / Magic Methods in Python

Dunder methods, also called magic methods, are special methods in Python whose names start and end with double underscores (`__`).

They allow Python objects to interact with built-in operations and functions.

Common examples include:

- `__init__()` — initializes an object when it is created.
- `__str__()` — defines the readable string representation of an object.
- `__len__()` — defines what should be returned when `len()` is used on an object.
- `__add__()` — defines behavior when two objects are added using `+`.

Dunder methods are usually called automatically by Python rather than directly by the programmer.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Student: {self.name}, Age: {self.age}"


student = Student("Areeba", 22)

print(student)
```

### Example of `__len__()`

The `__len__()` method defines what Python should return when the `len()` function is used with an object.

```python
class Course:
    def __init__(self, name, topics):
        self.name = name
        self.topics = topics

    def __len__(self):
        return len(self.topics)


course = Course(
    "Python Functions",
    ["Functions", "Methods", "Dunder Methods"]
)

print("Number of topics:", len(course))
```

### Example of `__add__()`

The `__add__()` method defines how the `+` operator should work with objects.

```python
class Score:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Score(self.value + other.value)

    def __str__(self):
        return str(self.value)


score1 = Score(80)
score2 = Score(15)

total = score1 + score2

print("Total Score:", total)
```

## 2. Difference Between Functions and Methods

### Function

A function is a reusable block of code that performs a specific task. It is defined using the `def` keyword and can be called independently.

Example:

`len()` is a built-in function that can be used with different objects.

### Method

A method is a function that is defined inside a class and is associated with an object or class. It is usually called using the object followed by a dot (`.`).

Example:

`append()` is a list method used to add an item to a list.

### Key Differences

| Function | Method |
|---|---|
| Can be defined independently | Usually defined inside a class |
| Called by its name | Usually called through an object or class |
| Does not necessarily belong to an object | Associated with an object or class |
| Example: `len(my_list)` | Example: `my_list.append(10)` |

```python
def calculate_average(numbers):
    total = sum(numbers)
    return total / len(numbers)


numbers = [10, 20, 30, 40, 50]

average = calculate_average(numbers)

print("Numbers:", numbers)
print("Average:", average)
```

### Practical Example of a Method

The `append()` operation is a list method because it belongs to a list object.

```python
students = ["Ali", "Sara"]

students.append("Areeba")

print("Students:", students)
```

```python
numbers = [10, 20, 30]

print("Length:", len(numbers))

numbers.append(40)

print("Updated list:", numbers)
```

## 3. Conclusion

Functions and methods both help make code reusable and organized. A function can work independently, while a method is associated with an object or class.

Dunder or magic methods are special methods that allow Python objects to work with built-in operations such as initialization, printing, length calculation, and addition.

Understanding these concepts helps in writing cleaner, reusable, and object-oriented Python programs.
