## What is Machine Learning?

Machine Learning (ML) is a branch of Artificial Intelligence that allows computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every possible situation.

Machine Learning can be broadly divided into three major types:

- Supervised Learning
- Unsupervised Learning
- Reinforcement Learning

## 1. Supervised Learning

Supervised Learning is a type of Machine Learning where a model learns from labeled data.

In labeled data, both the input and the correct output are provided to the model. The model learns the relationship between the input features and the target output and then uses this knowledge to make predictions on new data.

### Examples

- House price prediction
- Spam email detection
- Disease classification
- Student grade prediction

### Common Algorithms

- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)

### Supervised Learning Flow

```text
Labeled Training Data
        ↓
Input Features + Correct Labels
        ↓
    ML Algorithm
        ↓
    Model Training
        ↓
    Trained Model
        ↓
    New / Unseen Data
        ↓
      Prediction

````python
# Simple supervised learning example
# The model learns a relationship between input and output.

hours_studied = [2, 4, 6, 8]
exam_scores = [50, 60, 70, 80]

print("Study Hours:", hours_studied)
print("Exam Scores:", exam_scores)
````

## 2. Unsupervised Learning

Unsupervised Learning is a type of Machine Learning where the model learns from data without predefined labels.

The algorithm tries to discover hidden patterns, groups, or relationships within the data.

### Examples

- Customer segmentation
- Grouping similar products
- Finding unusual patterns
- Document clustering

### Common Algorithms

- K-Means Clustering
- Hierarchical Clustering
- DBSCAN
- Principal Component Analysis (PCA)

````python
# Example of unlabeled customer data
# There is no predefined target label.

customers = [
    [25, 200],
    [30, 250],
    [45, 900],
    [50, 1000]
]

print("Customer Data:")
for customer in customers:
    print(customer)
````

### Unsupervised Learning Flow

```text
Unlabeled Data
      ↓
   ML Algorithm
      ↓
Pattern Detection
      ↓
Grouping / Clustering
      ↓
Discovered Patterns
      ↓
Useful Insights


## 3. Reinforcement Learning

Reinforcement Learning is a type of Machine Learning where an agent learns by interacting with an environment.

The agent takes actions and receives rewards or penalties based on those actions. Over time, the agent learns a strategy that helps it maximize its total reward.

### Examples

- Game-playing AI
- Robot navigation
- Self-driving systems
- Recommendation systems

### Important Terms

- Agent: The learner or decision-maker
- Environment: The world in which the agent operates
- Action: A decision taken by the agent
- Reward: Feedback received after an action
- Policy: Strategy used by the agent

````python
# Simple reinforcement learning example

actions = ["Move Left", "Move Right"]
reward = 10

print("Available Actions:", actions)
print("Reward received:", reward)
````

###Reinforcement Learning Flow

```text
      Environment
          ↓
        State
          ↓
        Agent
          ↓
        Action
          ↓
      Environment
          ↓
    Reward / Penalty
          ↓
        Agent
          ↓
   Learns Better Policy

## 4. Regression vs Classification

Both Regression and Classification are supervised learning techniques.

### Regression

Regression predicts a continuous numerical value.

#### Examples

1. Predicting house prices
2. Predicting temperature
3. Predicting a person's salary

Example output:

- $250,000
- 32.5°C
- $60,000

### Classification

Classification predicts a category or class.

#### Examples

1. Spam vs Not Spam
2. Pass vs Fail
3. Disease vs No Disease

Example output:

- Spam
- Pass
- Disease

### Difference Between Regression and Classification

| Feature | Regression | Classification |
|---|---|---|
| Output | Continuous value | Category/Class |
| Example | House price | Spam detection |
| Prediction | 250000 | Spam |
| Type | Numerical | Categorical |

### Simple Rule

Regression → Predicts a number

Classification → Predicts a category

## 5. Decision Tree

A Decision Tree is a supervised Machine Learning algorithm that makes predictions by splitting data into different branches based on conditions.

It looks similar to a flowchart.

### Main Components

- Root Node: Starting point
- Decision Node: A condition/question
- Branch: Result of a condition
- Leaf Node: Final prediction

### Example

Suppose we want to decide whether a person should play outside.

The decision can depend on the weather.

````python
# Simple decision tree logic example

weather = "Sunny"

if weather == "Sunny":
    print("Play Outside")
else:
    print("Stay Inside")
````

# Task 2: Simple Decision Tree Simulator

## Objective

The purpose of this task is to simulate a simple decision tree for loan approval using Python conditional statements.

The program takes three inputs:

- Age
- Income
- Credit Score

It then follows a sequence of decision rules to determine whether the loan should be approved or rejected.

### Decision Rules

1. If age is less than 18 → Reject
2. If income is less than 30,000 → Reject
3. If credit score is less than 600 → Reject
4. Otherwise → Approve

The program also prints the decision path so that the user can see how the final decision was reached.

### How It Mimics a Decision Tree

The `if`, `elif`, and `else` statements represent decision nodes in a decision tree. Each condition checks one feature and sends the decision to the next branch. The final result, either Approved or Rejected, represents a leaf node.

````python
# Simple Decision Tree Simulator
# This program simulates a decision tree for loan approval.

age = int(input("Enter your age: "))
income = float(input("Enter your annual income: "))
credit_score = int(input("Enter your credit score: "))

print("\nDecision Path:")

# First decision node: check age.
if age < 18:
    print("Age < 18 → Reject")
    print("Loan Status: Rejected")

# Second decision node: check income.
elif income < 30000:
    print("Age >= 18 → Continue")
    print("Income < 30000 → Reject")
    print("Loan Status: Rejected")

# Third decision node: check credit score.
elif credit_score < 600:
    print("Age >= 18 → Continue")
    print("Income >= 30000 → Continue")
    print("Credit Score < 600 → Reject")
    print("Loan Status: Rejected")

# Final decision: approve the loan.
else:
    print("Age >= 18 → Continue")
    print("Income >= 30000 → Continue")
    print("Credit Score >= 600 → Approve")
    print("Loan Status: Approved")
````
