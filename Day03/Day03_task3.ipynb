## 3. Research from Gemini

### Question Asked

"What are best practices for if-elif-else statements in Python?"

### Key Findings from Gemini

Gemini explained conditional statements from both a computer science and practical
programming perspective.

#### 1. Control Flow

An `if-elif-else` structure changes the normal linear flow of a program by allowing
the program to choose different execution paths depending on conditions.

#### 2. Cyclomatic Complexity

Gemini explained that conditional branches increase cyclomatic complexity.
More branches mean more possible execution paths, which can make code harder to
test and maintain.

#### 3. Performance

An `if-elif` chain may evaluate conditions sequentially until a matching condition
is found. Therefore, a long chain can require more condition checks.

#### 4. Practical Coding Recommendations

Gemini highlighted several important practices:

- Use dictionary lookups when conditions are simple key-based selections.
- Use guard clauses when they make the logic clearer.
- Avoid unnecessary object-oriented designs just to remove a small conditional.
- Use `match-case` when it makes multi-way branching clearer.
- Use `if-elif-else` when conditions involve ranges or complex expressions.
- Consider readability and maintainability instead of blindly following one rule.
- Choose the conditional structure according to the problem.

### Gemini's Main Conclusion

Gemini emphasizes that there is no single conditional style that is best for
every situation. Simple `if-elif-else` statements are appropriate for range
checks and complex conditions, while dictionaries or `match-case` can be
useful for suitable multi-way selections.

The most important consideration is readable, maintainable, and appropriate code.

## 7. Comparison of Conditional Coding Styles

### Style 1: Separate `if` Statements

```python
score = 85

if score >= 90:
    grade = "A"

if score >= 80:
    grade = "B"

if score >= 70:
    grade = "C"

### Style 2: if-elif-else

````python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
````

### Style 3: Nested if

````python
score = 85

if score >= 0:
    if score >= 80:
        grade = "B"
    else:
        grade = "C"
````

### Style 4: Guard Clause

````python
def check_grade(score):
    if score < 0 or score > 100:
        return "Invalid grade"

    if score >= 90:
        return "A"

    if score >= 80:
        return "B"

    return "Below B"
````
