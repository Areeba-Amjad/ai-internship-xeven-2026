## 2. Accessing List Items - Indexing

Python lists use zero-based indexing.

This means:

- Index `0` → First item
- Index `1` → Second item
- Index `2` → Third item

Python also supports negative indexing:

- `-1` → Last item
- `-2` → Second-last item

````python
fruits = ["apple", "banana", "cherry", "date"]

print("First item:", fruits[0])
print("Second item:", fruits[1])
print("Last item:", fruits[-1])
print("Second-last item:", fruits[-2])
````

## 3. List Slicing

List slicing allows us to extract a portion of a list.

### Syntax

`list[start:stop:step]`

The `stop` index is not included.

Examples:

- `numbers[:5]` → First 5 elements
- `numbers[-5:]` → Last 5 elements
- `numbers[::2]` → Every second element
- `numbers[::-1]` → Reverse the entire list

````python
numbers = list(range(1, 21))

print("First 5:", numbers[:5])
print("Last 5:", numbers[-5:])
print("Every 3rd element:", numbers[::3])
print("Reversed:", numbers[::-1])
print("Middle 10:", numbers[5:15])
````

## 4. Adding Items to Lists

Python provides three important methods for adding items:

### `append()`
Adds one item to the end of a list.

### `insert()`
Adds an item at a specific index.

### `extend()`
Adds multiple items from another iterable.

````python
students = ["Ali", "Sara"]

# Add one item at the end
students.append("Ahmed")
print("After append():", students)

# Add an item at a specific position
students.insert(1, "Hina")
print("After insert():", students)

# Add multiple items
students.extend(["Usman", "Ayesha"])
print("After extend():", students)
````

## 5. Removing Items from Lists

### `remove()`
Removes the first occurrence of a specified value.

### `pop()`
Removes and returns an item by index. Without an index, it removes the last item.

### `clear()`
Removes all items from the list.

students = ["Ali", "Sara", "Ahmed", "Sara"]

students.remove("Sara")
print("After remove():", students)

removed_student = students.pop()
print("Removed student:", removed_student)
print("After pop():", students)

students.clear()
print("After clear():", students)

## 6. Sorting and Reversing Lists

### `sort()`
Sorts a list in place.

### `reverse()`
Reverses the order of items in place.

````python
scores = [45, 88, 12, 99, 32]

scores.sort()
print("Ascending:", scores)

scores.sort(reverse=True)
print("Descending:", scores)

scores.reverse()
print("Reversed:", scores)
````

## 7. Handling IndexError

An `IndexError` occurs when we try to access an index that does not exist in a list.

We can use `try-except` to handle this error gracefully.

````python
numbers = [10, 20, 30]

try:
    print(numbers[5])
except IndexError:
    
    print("Error: Index is out of range.")
````

## 9. Task 2 - Grade Tracker with Lists

This task uses two parallel lists:

- `student_names`
- `student_grades`

The program finds:

- Highest grade and student name.
- Lowest grade and student name.
- Average grade.
- Students who passed with a grade of 50 or above.

````python
# Create two parallel lists for student names and grades
student_names = ["Ali", "Ahmed", "Sara", "Hina", "Usman"]
student_grades = [85, 72, 91, 45, 68]

# Find the highest grade and its index
highest_grade = max(student_grades)
highest_index = student_grades.index(highest_grade)
highest_student = student_names[highest_index]

# Find the lowest grade and its index
lowest_grade = min(student_grades)
lowest_index = student_grades.index(lowest_grade)
lowest_student = student_names[lowest_index]

# Calculate the average grade
average_grade = sum(student_grades) / len(student_grades)

# Find students who passed
passed_students = []

for index in range(len(student_names)):
    if student_grades[index] >= 50:
        passed_students.append(student_names[index])

# Display the results
print("===== Grade Tracker =====")
print("Highest Grade:", highest_grade)
print("Highest Grade Student:", highest_student)

print("Lowest Grade:", lowest_grade)
print("Lowest Grade Student:", lowest_student)

print("Average Grade:", average_grade)

print("Passed Students:", passed_students)
````

## 10. Task 3 - List Slicing Practice

This task demonstrates different slicing patterns using numbers from 1 to 20:

- First 5 elements.
- Last 5 elements.
- Every 3rd element.
- Reverse of the entire list.
- Middle 10 elements.

````python
# Create a list of numbers from 1 to 20
numbers = list(range(1, 21))

print("Original List:", numbers)

# Get the first 5 elements
first_five = numbers[:5]
print("First 5 Elements:", first_five)

# Get the last 5 elements
last_five = numbers[-5:]
print("Last 5 Elements:", last_five)

# Get every 3rd element
every_third = numbers[::3]
print("Every 3rd Element:", every_third)

# Reverse the entire list using slicing
reversed_list = numbers[::-1]
print("Reversed List:", reversed_list)

# Get the middle 10 elements
middle_ten = numbers[5:15]
print("Middle 10 Elements:", middle_ten)
````

## 15. Performance Considerations

List performance becomes important when working with large amounts of data.

### `append()`

`append()` adds an item to the end of the list and is generally very efficient.

### `insert()`

`insert()` can be slower when inserting near the beginning or middle because existing elements may need to be shifted.

### `extend()`

`extend()` is useful for adding multiple items to a list and can be more appropriate than repeatedly calling `append()`.

### `remove()`

`remove()` searches for the value before removing it, so the operation can take longer for large lists.

### `pop()`

`pop()` without an index removes the last item and is generally efficient. Removing from the beginning can require shifting remaining elements.

### `sort()`

Sorting requires processing the list elements and therefore becomes more computationally significant as the list grows.

For very large datasets or specialized workloads, other data structures or libraries may be more suitable than Python lists.
