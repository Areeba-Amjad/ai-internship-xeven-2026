## 1. Logical Operators

Python has three main logical operators:

- `and` – True when all conditions are True.
- `or` – True when at least one condition is True.
- `not` – Reverses the Boolean value.

### Truth Table

| A | B | A and B | A or B |
|---|---|---------|--------|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

For `not`:

| A | not A |
|---|-------|
| True | False |
| False | True |

````python
age = 20
has_id = True

print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
````

## 5. Operator Precedence

Operator precedence determines the order in which Python evaluates
different operators in an expression.

Python generally follows this order:

1. Parentheses `()`
2. Exponentiation `**`
3. Multiplication `*`, Division `/`, Floor Division `//`,
   and Modulus `%`
4. Addition `+` and Subtraction `-`

Parentheses can be used to change the normal order of operations.

Understanding operator precedence helps avoid incorrect results
when writing mathematical expressions and calculations.

````python
# Operator Precedence

result_1 = 10 + 5 * 2
result_2 = (10 + 5) * 2
result_3 = 2 ** 3 * 2
result_4 = 20 / 5 + 3

print("10 + 5 * 2 =", result_1)
print("(10 + 5) * 2 =", result_2)
print("2 ** 3 * 2 =", result_3)
print("20 / 5 + 3 =", result_4)
````

## 6. Common Type Conversion Errors

Type conversion can cause errors when a value cannot be converted
into the required data type.

For example, converting non-numeric text into an integer using
`int()` will produce a `ValueError`.

The `try-except` statement can be used to handle such errors
without stopping the entire program.

````python
# Handling a type conversion error

user_age = input("Enter your age: ")

try:
    age = int(user_age)
    print("Your age is:", age)
except ValueError:
    print("Error: Please enter a valid number for age.")
````

## 8. Advanced Login System

The Advanced Login System validates three conditions:

1. Username must be at least 5 characters long.
2. Password must be at least 8 characters long.
3. Age must be 18 or above.

Logical operators and comparison operators are used to check
whether all requirements are satisfied.

If all conditions are met, access is granted. Otherwise,
access is denied and specific error messages are displayed.

````python
# Advanced Login System

# Get user information
username = input("Enter your username: ")
password = input("Enter your password: ")
age_input = input("Enter your age: ")

# Convert age from string to integer
age = int(age_input)

# Validate username
if len(username) < 5:
    print("Error: Username must be at least 5 characters long.")

# Validate password
if len(password) < 8:
    print("Error: Password must be at least 8 characters long.")

# Validate age
if age < 18:
    print("Error: You must be 18 or above to access the system.")

# Check all conditions using the logical operator 'and'
if (
    len(username) >= 5
    and len(password) >= 8
    and age >= 18
):
    print("Access granted! Login successful.")
else:
    print("Access denied. Please fix the above errors.")
````

##  Task 2: Multi-Operation Calculator

The Multi-Operation Calculator accepts two numbers and an arithmetic
operation from the user. The program converts the input values from
strings to floating-point numbers using `float()`.

It supports the following operations:

- Addition `+`
- Subtraction `-`
- Multiplication `*`
- Division `/`
- Modulus `%`
- Exponentiation `**`

The program also handles division by zero and invalid operation
selections. Float results are formatted to one decimal place.

````python
# Task 2: Multi-Operation Calculator

number_1 = float(input("Enter first number: "))
number_2 = float(input("Enter second number: "))

operation = input("Enter operation (+, -, *, /, %, **): ")

if operation == "+":
    result = number_1 + number_2
    print(f"{number_1:.1f} + {number_2:.1f} = {result:.1f}")

elif operation == "-":
    result = number_1 - number_2
    print(f"{number_1:.1f} - {number_2:.1f} = {result:.1f}")

elif operation == "*":
    result = number_1 * number_2
    print(f"{number_1:.1f} * {number_2:.1f} = {result:.1f}")

elif operation == "/":
    if number_2 == 0:
        print("Error: Division by zero is not allowed.")
    else:
        result = number_1 / number_2
        print(f"{number_1:.1f} / {number_2:.1f} = {result:.1f}")

elif operation == "%":
    if number_2 == 0:
        print("Error: Modulus by zero is not allowed.")
    else:
        result = number_1 % number_2
        print(f"{number_1:.1f} % {number_2:.1f} = {result:.1f}")

elif operation == "**":
    result = number_1 ** number_2
    print(f"{number_1:.1f} ** {number_2:.1f} = {result:.1f}")

else:
    print("Error: Invalid operation. Please choose +, -, *, /, %, or **.")
````
