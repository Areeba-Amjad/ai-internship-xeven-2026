## 1. What is a Vector Database?

A vector database is a specialized database designed to store, index, and search high-dimensional vectors.

Vectors are usually generated from data such as:

- Text
- Images
- Audio
- Documents

For example, an embedding may look like:

[0.12, -0.31, 0.88, 0.45, ...]

The main purpose of a vector database is to find vectors that are semantically similar to a query vector.

### Basic Workflow

1. Convert documents into embeddings
2. Store embeddings in a vector database
3. Convert the user query into an embedding
4. Search for similar vectors
5. Return the Top-K results

## 2. Why Not Traditional Databases?

Traditional databases are very good for:

- Exact matching
- Structured data
- SQL queries
- Joins
- Sorting
- Aggregation

However, semantic similarity search requires comparing high-dimensional vectors.

A simple approach would compare a query vector with every stored vector. This is called exact or brute-force search.

For large datasets, this can become computationally expensive.

Vector databases use Approximate Nearest Neighbor (ANN) algorithms to search for similar vectors much faster.

### Trade-off

Exact Search:
- Higher accuracy
- Can be slower for large datasets

ANN Search:
- Much faster
- May have slightly lower recall

## 3. Approximate Nearest Neighbor (ANN)

ANN stands for Approximate Nearest Neighbor.

The goal is to find vectors that are very close to a query vector without comparing the query against every single vector.

ANN algorithms improve:

- Search speed
- Scalability
- Query latency

The trade-off is that ANN may not always return the exact nearest neighbors.

Therefore, ANN systems are usually evaluated using recall, latency, and throughput.

## 4. Vector Indexing Algorithms

### HNSW

HNSW stands for Hierarchical Navigable Small World.

It organizes vectors into a graph with multiple layers.

The upper layers allow fast navigation, while lower layers provide more detailed searching.

Advantages:

- Fast similarity search
- Good recall
- Low latency

Trade-off:

- Requires additional memory
- Index construction can take time


### IVF

IVF stands for Inverted File Index.

IVF divides vectors into different clusters.

During search, instead of checking every vector, only selected clusters are searched.

Advantages:

- Faster search on large datasets
- Reduces the number of vectors that need to be examined

Trade-off:

- Search quality depends on clustering and number of clusters searched

## 5. Vector Database Options

| Vector Store | Type | Main Use |
|---|---|---|
| FAISS | Local Library | Fast local similarity search |
| Pinecone | Managed Cloud | Production cloud applications |
| Qdrant | Open Source | Production and self-hosted applications |

### FAISS

FAISS is developed for efficient similarity search and clustering of dense vectors.

### Pinecone

Pinecone is a managed cloud vector database designed for production AI applications.

### Qdrant

Qdrant is an open-source vector database that supports vector similarity search and metadata filtering.

```python
np.random.seed(42)

num_vectors = 10000
dimension = 128

embeddings = np.random.random(
    (num_vectors, dimension)
).astype("float32")

print("Embeddings shape:", embeddings.shape)
```

```python
index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

print("Vectors stored:", index.ntotal)
```

```python
query = np.random.random(
    (1, dimension)
).astype("float32")

k = 5

distances, indices = index.search(query, k)

print("Top 5 similar vectors:")
print(indices)

print("\nDistances:")
print(distances)
```

```python
hnsw_index = faiss.IndexHNSWFlat(
    dimension,
    32
)

hnsw_index.hnsw.efConstruction = 40
hnsw_index.hnsw.efSearch = 32

hnsw_index.add(embeddings)

print("HNSW vectors:", hnsw_index.ntotal)
```

```python
hnsw_distances, hnsw_indices = hnsw_index.search(
    query,
    k
)

print("HNSW Top-5 results:")
print(hnsw_indices)

print("\nDistances:")
print(hnsw_distances)
```

```python
nlist = 100

quantizer = faiss.IndexFlatL2(dimension)

ivf_index = faiss.IndexIVFFlat(
    quantizer,
    dimension,
    nlist,
    faiss.METRIC_L2
)

print("IVF index created ✅")
```

```python
ivf_index.train(embeddings)

print("IVF trained:", ivf_index.is_trained)
```

```python
ivf_index.add(embeddings)

print("IVF vectors:", ivf_index.ntotal)
```

```python
ivf_index.nprobe = 10

ivf_distances, ivf_indices = ivf_index.search(
    query,
    k
)

print("IVF Top-5 results:")
print(ivf_indices)

print("\nDistances:")
print(ivf_distances)
```

```python
exact_distances, exact_indices = index.search(
    query,
    k
)

hnsw_distances, hnsw_indices = hnsw_index.search(
    query,
    k
)

ivf_distances, ivf_indices = ivf_index.search(
    query,
    k
)

def recall_at_k(exact, approximate):
    exact_set = set(exact[0])
    approximate_set = set(approximate[0])
    
    return len(
        exact_set.intersection(approximate_set)
    ) / len(exact_set)


hnsw_recall = recall_at_k(
    exact_indices,
    hnsw_indices
)

ivf_recall = recall_at_k(
    exact_indices,
    ivf_indices
)

print(f"HNSW Recall@{k}: {hnsw_recall:.2%}")
print(f"IVF Recall@{k}: {ivf_recall:.2%}")
```

```python
import time

def measure_latency(index, query, k, repetitions=100):
    start = time.perf_counter()

    for _ in range(repetitions):
        index.search(query, k)

    elapsed = time.perf_counter() - start

    return (elapsed / repetitions) * 1000


flat_latency = measure_latency(
    index,
    query,
    k
)

hnsw_latency = measure_latency(
    hnsw_index,
    query,
    k
)

ivf_latency = measure_latency(
    ivf_index,
    query,
    k
)

print(f"Flat Latency : {flat_latency:.4f} ms")
print(f"HNSW Latency : {hnsw_latency:.4f} ms")
print(f"IVF Latency  : {ivf_latency:.4f} ms")
```

```python
def measure_throughput(index, query, k, repetitions=100):
    start = time.perf_counter()

    for _ in range(repetitions):
        index.search(query, k)

    elapsed = time.perf_counter() - start

    return repetitions / elapsed


flat_qps = measure_throughput(
    index,
    query,
    k
)

hnsw_qps = measure_throughput(
    hnsw_index,
    query,
    k
)

ivf_qps = measure_throughput(
    ivf_index,
    query,
    k
)

print(f"Flat QPS : {flat_qps:.2f}")
print(f"HNSW QPS : {hnsw_qps:.2f}")
print(f"IVF QPS  : {ivf_qps:.2f}")
```

## Task 2 — Document Library with FAISS

```python
topics = {
    "AI": [
        "Artificial intelligence enables machines to perform intelligent tasks.",
        "AI systems can analyze data and make predictions.",
        "Artificial intelligence is used in healthcare, finance, and education."
    ],
    
    "Machine Learning": [
        "Machine learning algorithms learn patterns from training data.",
        "Supervised learning uses labeled training examples.",
        "Unsupervised learning discovers hidden patterns in data."
    ],
    
    "Data Science": [
        "Data science combines statistics, programming, and domain knowledge.",
        "Data scientists analyze datasets to discover useful insights.",
        "Data visualization helps communicate patterns and trends."
    ],
    
    "Databases": [
        "Databases organize and store structured information.",
        "SQL databases use tables and relationships.",
        "Database indexing improves query performance."
    ],
    
    "Cloud Computing": [
        "Cloud computing provides computing resources over the internet.",
        "Cloud platforms provide scalable storage and computing.",
        "Cloud services reduce the need for local infrastructure."
    ],
    
    "Cyber Security": [
        "Cybersecurity protects systems and data from unauthorized access.",
        "Encryption protects sensitive information.",
        "Authentication verifies the identity of users."
    ]
}

library = []

doc_id = 1

for topic, texts in topics.items():
    for text in texts:
        for version in range(1, 4):
            library.append({
                "id": doc_id,
                "text": text,
                "source": f"{topic.lower().replace(' ', '_')}_document_{version}.txt",
                "page": version,
                "section": topic
            })
            doc_id += 1

print("Total documents:", len(library))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=100
)

chunks = []

for doc in library:
    doc_chunks = splitter.split_text(doc["text"])
    
    for chunk_number, chunk in enumerate(doc_chunks, start=1):
        chunks.append({
            "text": chunk,
            "source": doc["source"],
            "page": doc["page"],
            "section": doc["section"],
            "chunk": chunk_number
        })

print("Total chunks:", len(chunks))
```

```python
chunk_texts = [item["text"] for item in chunks]

library_vectorizer = TfidfVectorizer()

library_embeddings = library_vectorizer.fit_transform(
    chunk_texts
).toarray().astype("float32")

print("Embedding shape:", library_embeddings.shape)
```

```python
library_index = faiss.IndexFlatL2(
    library_embeddings.shape[1]
)

library_index.add(library_embeddings)

print("Library vectors stored:", library_index.ntotal)
```

```python
def search_library(query, k=5):
    
    query_vector = library_vectorizer.transform(
        [query]
    ).toarray().astype("float32")
    
    distances, indices = library_index.search(
        query_vector,
        k
    )
    
    print("Query:", query)
    print("=" * 70)
    
    for rank, idx in enumerate(indices[0], start=1):
        item = chunks[idx]
        
        print(f"\nResult {rank}")
        print("Text:", item["text"])
        print("Source:", item["source"])
        print("Page:", item["page"])
        print("Section:", item["section"])
        print("Distance:", distances[0][rank - 1])
```

```python
search_library(
    "How does machine learning work?",
    k=5
)
```

```python
def filtered_search(query, section=None, source=None, k=5):
    
    query_vector = library_vectorizer.transform(
        [query]
    ).toarray().astype("float32")
    
    distances, indices = library_index.search(
        query_vector,
        library_index.ntotal
    )
    
    results = []
    
    for distance, idx in zip(distances[0], indices[0]):
        item = chunks[idx]
        
        if section and item["section"] != section:
            continue
        
        if source and item["source"] != source:
            continue
        
        results.append((distance, item))
        
        if len(results) == k:
            break
    
    print("Query:", query)
    print("=" * 70)
    
    for rank, (distance, item) in enumerate(results, start=1):
        print(f"\nResult {rank}")
        print("Text:", item["text"])
        print("Source:", item["source"])
        print("Page:", item["page"])
        print("Section:", item["section"])
        print("Distance:", distance)
```

```python
filtered_search(
    "learning algorithms",
    section="Machine Learning",
    k=5
)
```

```python
import faiss
import numpy as np
import time
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample documents
documents = [
    "Machine learning allows computers to learn patterns from data.",
    "Artificial intelligence enables machines to perform intelligent tasks.",
    "Deep learning uses neural networks with multiple layers.",
    "Data science combines statistics, programming, and domain knowledge.",
    "Vector databases store and search high-dimensional vectors.",
    "FAISS provides fast similarity search for numerical vectors.",
    "Chroma is a persistent vector database for AI applications.",
    "Natural language processing helps computers understand human language.",
    "Computer vision enables machines to analyze images.",
    "Retrieval augmented generation combines retrieval with language generation."
]

# Create vectors
vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(documents).toarray().astype("float32")

# Create FAISS index
index = faiss.IndexFlatL2(vectors.shape[1])
index.add(vectors)

# Query
query = "How does machine learning work?"

query_vector = vectorizer.transform(
    [query]
).toarray().astype("float32")

# Measure latency
start = time.perf_counter()

distances, indices = index.search(
    query_vector,
    5
)

faiss_latency = (time.perf_counter() - start) * 1000

print("Query:", query)
print("FAISS latency:", round(faiss_latency, 4), "ms")
print("Top 5 results:")

for rank, idx in enumerate(indices[0], start=1):
    print(f"{rank}. {documents[idx]}")
```

```python
print("=" * 50)
print("FAISS vs CHROMA - FINAL RESULTS")
print("=" * 50)

print(f"FAISS  Query Latency : {faiss_latency:.4f} ms")
print(f"Chroma Query Latency : {chroma_latency:.4f} ms")

if faiss_latency < chroma_latency:
    print("\nFAISS was faster in this benchmark ✅")
else:
    print("\nChroma was faster in this benchmark ✅")
```
