# 1. What are Large Language Models (LLMs)?

Large Language Models (LLMs) are advanced Artificial Intelligence models trained on very large amounts of text data. They are designed to understand, process, and generate human-like language.

LLMs are commonly based on the **Transformer architecture**, which uses an **attention mechanism** to understand relationships between words and tokens in a sequence.

## Transformer Architecture

The Transformer architecture was introduced in the research paper **"Attention Is All You Need"**. Unlike older sequence models such as RNNs, Transformers can process many tokens in parallel.

The main components include:

- **Input Embeddings:** Convert tokens into numerical vectors.
- **Positional Encoding:** Provides information about the position of tokens.
- **Self-Attention:** Determines which tokens are important in relation to other tokens.
- **Feed-Forward Networks:** Further process the information.
- **Encoder and Decoder:** Used depending on the Transformer architecture and task.

## Attention Mechanism

Attention allows a model to focus on the most relevant parts of the input when processing a particular token.

For example, in the sentence:

> "The cat sat on the mat because it was tired."

The attention mechanism helps the model understand that **"it"** most likely refers to **"the cat"**.

This allows LLMs to understand relationships between words even when they are far apart in a sentence.

## Pre-training vs Fine-tuning

### Pre-training

Pre-training is the initial stage where an LLM learns general language patterns from a very large dataset.

During pre-training, the model learns:

- Grammar
- Vocabulary
- Language patterns
- General knowledge
- Relationships between words and concepts

### Fine-tuning

Fine-tuning is performed after pre-training. The model is trained further on a smaller, task-specific dataset.

For example, an LLM can be fine-tuned for:

- Customer support
- Medical text classification
- Code generation
- Sentiment analysis
- Question answering

### Summary

Pre-training provides **general language capabilities**, while fine-tuning adapts the model for **specific tasks or domains**.

# 3. Key Concepts: Tokens, Context Window, and Temperature

Understanding tokens, context windows, and temperature is important when working with LLM APIs.

## Tokens

A token is a unit of text processed by an LLM.

A token can represent:

- A complete word
- Part of a word
- Punctuation
- Numbers
- Special characters

For example, a sentence is converted into tokens before being processed by the model.

LLM APIs generally calculate input and output usage based on tokens.

## Context Window

The context window is the maximum amount of information that a model can process within a single interaction.

It can include:

- Previous conversation messages
- User prompts
- System instructions
- Documents
- Model-generated responses

A larger context window allows the model to work with longer documents and conversations.

However, the context window is not unlimited. If the amount of input becomes larger than the supported context window, some information may need to be removed or summarized.

## Temperature

Temperature controls the randomness of the model's output.

### Low Temperature

A low temperature generally produces:

- More predictable responses
- More consistent wording
- Less variation

It is useful for tasks such as:

- Classification
- Structured responses
- Factual extraction

### High Temperature

A higher temperature generally produces:

- More varied responses
- More creative wording
- Greater randomness

It can be useful for:

- Creative writing
- Brainstorming
- Generating multiple ideas

### Important Note

Temperature does not directly make a model "more intelligent." It mainly changes the randomness of token selection during generation.

# 4. Important LLM API Parameters

LLM APIs provide parameters that control how the model processes prompts and generates responses.

## max_tokens

`max_tokens` controls the maximum amount of output the model can generate.

A larger value allows longer responses, while a smaller value limits the response length.

> Note: The exact parameter name and behavior can vary between providers and model APIs.

## temperature

`temperature` controls the randomness of generated output.

- Lower value → more predictable output
- Higher value → more varied output

## top_p

`top_p` is a sampling parameter that limits token selection to a probability mass.

For example, a smaller `top_p` restricts the model to a smaller set of likely tokens, while a larger value allows more possibilities.

`temperature` and `top_p` both influence randomness, so they are generally tuned carefully rather than being changed aggressively at the same time.

## frequency_penalty

`frequency_penalty` reduces the likelihood of repeatedly using tokens that have already appeared frequently in the generated text.

It can help reduce repetitive responses.

## presence_penalty

`presence_penalty` encourages the model to introduce new topics or tokens rather than repeatedly using the same ones.

It can be useful when generating diverse content.

## Parameter Summary

| Parameter | Purpose |
|---|---|
| `max_tokens` | Controls maximum output length |
| `temperature` | Controls randomness |
| `top_p` | Controls probability-based token sampling |
| `frequency_penalty` | Reduces repeated token usage |
| `presence_penalty` | Encourages new topics/tokens |

Different LLM providers may use different parameter names, supported ranges, or API behavior. Developers should always check the documentation for the specific model being used.

# 6. Limitations of Large Language Models

Although LLMs are powerful, they have several important limitations.

## 1. Hallucinations

An LLM may generate information that sounds convincing but is incorrect or unsupported.

This is known as a **hallucination**.

Therefore, important information should be verified using reliable sources.

## 2. Knowledge Cutoff

A model may have limitations regarding information from events or developments that occurred after its training data or knowledge boundary.

For applications requiring current information, external tools such as search, databases, or retrieval systems may be required.

## 3. Context Limits

Every model has a maximum context window.

If a conversation or document exceeds this limit, the complete information may not be available to the model at once.

## 4. Reasoning Errors

LLMs can make mistakes in:

- Mathematical calculations
- Logical reasoning
- Multi-step problems
- Code
- Factual interpretation

Even when the response appears confident, it may still be incorrect.

## 5. Bias

LLMs learn patterns from their training data. If training data contains biases, the model may reproduce or amplify some of those biases.

## 6. Dependency on Prompt Quality

The quality and clarity of the prompt can significantly affect the output.

Poorly written prompts may result in:

- Ambiguous answers
- Missing information
- Incorrect assumptions
- Unwanted output formats

## Conclusion

LLMs are powerful tools but should not always be treated as completely reliable sources of truth. For high-impact applications, their outputs should be validated, monitored, and combined with reliable data sources or human review.

## Task 3 — Build Simple Chatbot

````python
# Task 3: Offline Simple Chatbot
# No API or billing required.

print("=" * 50)
print("       OFFLINE AI CHATBOT")
print("=" * 50)
print("Type 'exit' to stop the chatbot.")

while True:
    user_input = input("\nYou: ")

    if user_input.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    if not user_input.strip():
        print("Chatbot: Please enter a message.")
        continue

    # Simple offline responses
    user_lower = user_input.lower()

    if "python" in user_lower:
        response = (
            "Python is a popular programming language used in "
            "Data Science, AI, web development, and automation."
        )

    elif "llm" in user_lower:
        response = (
            "LLM stands for Large Language Model. "
            "LLMs are AI models trained on large amounts of text."
        )

    elif "hello" in user_lower or "hi" in user_lower:
        response = "Hello! How can I help you today?"

    elif "data science" in user_lower:
        response = (
            "Data Science combines statistics, programming, "
            "machine learning, and data analysis."
        )

    else:
        response = (
            "I am an offline demonstration chatbot. "
            "I can answer basic questions about Python, LLMs, "
            "and Data Science."
        )

    print("Chatbot:", response)
````

````python
# Define chatbot personality

system_message = """
You are a friendly and beginner-friendly Python and Data Science tutor.
You explain technical concepts using simple language and short examples.
You encourage students while helping them understand programming concepts.
"""

print("Chatbot Role:")
print(system_message)
````

````python
# Offline chatbot with conversation history

conversation_history = []

print("=" * 50)
print("     CONTEXT-AWARE OFFLINE CHATBOT")
print("=" * 50)
print("Type 'exit' to stop.")

while True:
    user_input = input("\nYou: ")

    if user_input.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    if not user_input.strip():
        print("Chatbot: Please enter a message.")
        continue

    # Store user message
    conversation_history.append({
        "role": "user",
        "content": user_input
    })

    # Generate offline response
    user_lower = user_input.lower()

    if "python" in user_lower:
        assistant_reply = (
            "Python is widely used in Data Science because "
            "of libraries such as NumPy, Pandas and Scikit-learn."
        )

    elif "llm" in user_lower:
        assistant_reply = (
            "LLMs are Large Language Models that process "
            "and generate human-like text."
        )

    elif "what did i ask" in user_lower:
        previous_questions = [
            message["content"]
            for message in conversation_history[:-1]
            if message["role"] == "user"
        ]

        if previous_questions:
            assistant_reply = (
                "Earlier you asked: "
                + " | ".join(previous_questions)
            )
        else:
            assistant_reply = "This is your first question."

    else:
        assistant_reply = (
            "I understand your question. "
            "This is an offline chatbot demonstration."
        )

    # Store assistant response
    conversation_history.append({
        "role": "assistant",
        "content": assistant_reply
    })

    print("Chatbot:", assistant_reply)
````

````python
# Error handling demonstration

def safe_chatbot_response(user_input):
    try:
        if not isinstance(user_input, str):
            raise TypeError("Input must be text.")

        if not user_input.strip():
            return "Please enter a valid message."

        return "Your message was received successfully."

    except TypeError as error:
        return f"Input Error: {error}"

    except Exception as error:
        return f"Unexpected Error: {error}"


# Test the function
print(safe_chatbot_response("Hello"))
print(safe_chatbot_response(""))
````
