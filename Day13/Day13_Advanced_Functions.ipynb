## 1. *args and **kwargs

### Definition

`*args` allows a function to accept a variable number of positional arguments.

`**kwargs` allows a function to accept a variable number of keyword arguments.

### When to Use

- Use `*args` when the number of positional arguments is unknown.
- Use `**kwargs` when the number of keyword arguments is unknown.
- They are useful when creating flexible and reusable functions.

### Unpacking

The `*` operator can unpack elements from a list or tuple into function arguments.

The `**` operator can unpack key-value pairs from a dictionary into keyword arguments.

### ChatGPT Research

ChatGPT explains that `*args` collects multiple positional arguments into a tuple, while `**kwargs` collects multiple keyword arguments into a dictionary.

`*args` is useful when we do not know how many positional values will be passed to a function.

`**kwargs` is useful when we want to pass a flexible number of named arguments.

Both can also be used for unpacking existing collections.

### Gemini Research

Gemini explains that `*args` provides flexibility for functions that can receive different numbers of positional arguments.

`**kwargs` provides flexibility for functions that receive named arguments.

The two techniques are commonly used in reusable functions, wrappers, and APIs.

### Claude Research

Claude describes `*args` as a mechanism for collecting positional arguments into a tuple and `**kwargs` as a mechanism for collecting keyword arguments into a dictionary.

They help developers create functions that can handle different numbers of inputs without explicitly defining every parameter.

## Task 1: Flexible Logger Function

Create a flexible logger function using `*args` and `**kwargs`.

The function accepts:
- A log level
- Multiple messages
- Optional timestamp
- Optional output file
- Optional output format

Different log levels will have different terminal colors.

```python
import json
from datetime import datetime


def log(level, *messages, **options):
    """Create a flexible logger with multiple messages and options."""
    timestamp = options.get("timestamp", True)
    file_name = options.get("file", "log.txt")
    output_format = options.get("format", "text")

    message = " ".join(str(item) for item in messages)

    if timestamp:
        current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    else:
        current_time = None

    log_data = {
        "level": level.upper(),
        "message": message,
    }

    if current_time:
        log_data["timestamp"] = current_time

    if output_format.lower() == "json":
        output = json.dumps(log_data)
    else:
        if current_time:
            output = f"[{current_time}] {level.upper()}: {message}"
        else:
            output = f"{level.upper()}: {message}"

    print(output)

    with open(file_name, "a", encoding="utf-8") as file:
        file.write(output + "\n")

    return output
```

```python
import json
from datetime import datetime


def log(level, *messages, **options):
    """Create a flexible logger with colors and multiple output options."""
    timestamp = options.get("timestamp", True)
    file_name = options.get("file", "log.txt")
    output_format = options.get("format", "text")

    message = " ".join(str(item) for item in messages)
    level = level.upper()

    if timestamp:
        current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    else:
        current_time = None

    log_data = {
        "level": level,
        "message": message,
    }

    if current_time:
        log_data["timestamp"] = current_time

    if output_format.lower() == "json":
        output = json.dumps(log_data)
    else:
        if current_time:
            output = f"[{current_time}] {level}: {message}"
        else:
            output = f"{level}: {message}"

    colors = {
        "ERROR": "\033[91m",
        "WARNING": "\033[93m",
        "INFO": "\033[0m",
    }

    color = colors.get(level, "\033[0m")
    reset = "\033[0m"

    print(f"{color}{output}{reset}")

    with open(file_name, "a", encoding="utf-8") as file:
        file.write(output + "\n")

    return output
```

```python
log("INFO", "Application started")
log("WARNING", "Memory usage is high")
log("ERROR", "Unable to connect to server")
```

## Task 2: Data Transformer Suite

### 2.1 Lambda with map()

Use `map()` and lambda to remove extra whitespace and convert strings to uppercase.

```python
names = [
    "  areeba  ",
    " ali ",
    " sara",
    " ahmed  "
]

cleaned_names = list(
    map(lambda name: name.strip().upper(), names)
)

print("Original:", names)
print("Transformed:", cleaned_names)
```

```python
import re


texts = [
    "Contact us at info@example.com",
    "Call us at 03001234567",
    "Visit https://example.com",
    "Hello everyone",
    "Email: student@gmail.com",
    "Phone: 03111234567",
    "Website: https://python.org"
]

emails = list(
    filter(
        lambda text: re.search(
            r"[\w.-]+@[\w.-]+\.\w+",
            text
        ),
        texts
    )
)

phone_numbers = list(
    filter(
        lambda text: re.search(
            r"\b03\d{9}\b",
            text
        ),
        texts
    )
)

urls = list(
    filter(
        lambda text: re.search(
            r"https?://\S+",
            text
        ),
        texts
    )
)

print("Emails:", emails)
print("Phone Numbers:", phone_numbers)
print("URLs:", urls)
```

### 2.3 Lambda with sorted()

Use `sorted()` and lambda to sort dictionaries using multiple keys.

The data will first be sorted by department and then by age.

```python
students = [
    {"name": "Areeba", "department": "Data Science", "age": 23},
    {"name": "Ali", "department": "Computer Science", "age": 21},
    {"name": "Sara", "department": "Data Science", "age": 21},
    {"name": "Ahmed", "department": "Computer Science", "age": 22},
]

sorted_students = sorted(
    students,
    key=lambda student: (
        student["department"],
        student["age"]
    )
)

for student in sorted_students:
    print(student)
```

### 2.4 Performance Comparison

The performance of lambda with `map()`, a regular function with `map()`, and a list comprehension will be compared using `timeit`.

The operation will square numbers from a large list.

```python
import timeit


numbers = list(range(10000))


def square(number):
    """Return the square of a number."""
    return number ** 2


lambda_time = timeit.timeit(
    lambda: list(map(lambda x: x ** 2, numbers)),
    number=100
)

function_time = timeit.timeit(
    lambda: list(map(square, numbers)),
    number=100
)

comprehension_time = timeit.timeit(
    lambda: [number ** 2 for number in numbers],
    number=100
)

print("Lambda + map:", lambda_time)
print("Regular function + map:", function_time)
print("List comprehension:", comprehension_time)
```

### Performance Observation

The execution time can vary depending on the computer and Python version.

List comprehensions are often very efficient for simple transformations. Lambda with `map()` and regular functions provide alternative approaches, but readability should also be considered when choosing an approach.

## Task 3: Advanced List/Dictionary Comprehensions

### 3.1 Flatten Nested Lists

Use a nested list comprehension to convert a nested list into a single flat list.

```python
nested = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

flattened = [
    item
    for sublist in nested
    for item in sublist
]

print("Nested list:", nested)
print("Flattened list:", flattened)
```

### 3.2 Matrix Transpose

A matrix transpose converts rows into columns and columns into rows.

This can be achieved using a nested list comprehension.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transpose = [
    [row[index] for row in matrix]
    for index in range(len(matrix[0]))
]

print("Original matrix:")
for row in matrix:
    print(row)

print("\nTransposed matrix:")
for row in transpose:
    print(row)
```

### 3.3 Dictionary Inversion

Use dictionary comprehension to swap dictionary keys and values.

```python
student_ids = {
    "Areeba": 101,
    "Ali": 102,
    "Sara": 103,
    "Ahmed": 104
}

inverted = {
    value: key
    for key, value in student_ids.items()
}

print("Original dictionary:", student_ids)
print("Inverted dictionary:", inverted)
```

### 3.4 Word Frequency Counter

Create a word frequency counter using dictionary comprehension.

The dictionary will contain each unique word as a key and its frequency as the value.

```python
text = """
python is easy
python is powerful
python is popular
"""

words = text.lower().split()

unique_words = set(words)

word_frequency = {
    word: words.count(word)
    for word in unique_words
}

print("Word Frequency:")
for word, frequency in sorted(word_frequency.items()):
    print(f"{word}: {frequency}")
```
