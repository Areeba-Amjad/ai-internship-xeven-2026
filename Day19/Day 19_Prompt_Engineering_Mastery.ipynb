## 1. What is Prompt Engineering?

Prompt engineering is the process of designing clear and structured instructions for Large Language Models (LLMs) to produce useful, accurate, and consistent outputs.

A good prompt can specify:

- The role of the AI
- The task to perform
- Relevant context
- Examples
- Expected output format
- Constraints and rules

Prompt engineering is especially important when building reliable AI applications.

## 2. Zero-Shot Prompting

Zero-shot prompting means asking an AI model to perform a task without providing examples.

The prompt contains only the task description and instructions.

### Example

Task:
Classify the following review as Positive, Negative, or Neutral.

Review:
"The product arrived on time and works perfectly."

Expected output:
Positive

### When to use Zero-Shot Prompting

Zero-shot prompting is suitable for:

- Simple classification
- Translation
- Summarization
- General questions
- Straightforward text transformation

It works best when the task is familiar and the expected behavior is clear.

```python
zero_shot_prompt = """
Classify the following customer review as Positive, Negative, or Neutral.

Review:
"The product arrived on time and works perfectly."

Return only the classification.
"""

print(zero_shot_prompt)
```

### Expected Result

Positive

The prompt does not provide any examples. It simply explains the task and specifies the required output.

## 3. Few-Shot Prompting

Few-shot prompting provides a small number of examples before asking the model to process a new input.

Usually, 2–5 examples can be enough to demonstrate the expected pattern.

### Example

Review: "I love this product."
Sentiment: Positive

Review: "The product stopped working."
Sentiment: Negative

Review: "The product is okay."
Sentiment: Neutral

Review: "Excellent quality and fast delivery."
Sentiment:

```python
few_shot_prompt = """
Classify each customer review as Positive, Negative, or Neutral.

Example 1:
Review: "I love this product."
Sentiment: Positive

Example 2:
Review: "The product stopped working."
Sentiment: Negative

Example 3:
Review: "The product is okay."
Sentiment: Neutral

Now classify:

Review: "Excellent quality and fast delivery."
Sentiment:
"""

print(few_shot_prompt)
```

### Expected Result

Positive

### Why Few-Shot is Better Here

The examples teach the model:

1. What the categories mean
2. What the expected output looks like
3. How the task should be performed

Few-shot prompting is especially useful for specialized tasks, custom categories, and strict output formats.

```python
comparison = {
    "Zero-Shot": {
        "Examples": "No examples",
        "Best For": "Simple and familiar tasks",
        "Advantage": "Short and efficient",
        "Limitation": "Model may interpret the desired format differently"
    },
    "Few-Shot": {
        "Examples": "2–5 examples",
        "Best For": "Specialized or structured tasks",
        "Advantage": "Demonstrates the desired pattern",
        "Limitation": "Uses more prompt tokens"
    }
}

for method, details in comparison.items():
    print(f"\n{method}")
    for key, value in details.items():
        print(f"{key}: {value}")
```

### Key Difference

Zero-shot = Tell the model what to do.

Few-shot = Tell the model what to do + Show examples.

Few-shot prompting is particularly useful when the task has unusual categories, terminology, or output requirements.

## 4. Chain-of-Thought (CoT) Prompting

Chain-of-Thought prompting is a technique that encourages a model to approach a complex problem through intermediate reasoning steps.

A simple form is:

"Let's think step by step."

CoT is particularly useful for:

- Mathematical problems
- Logical reasoning
- Multi-step decisions
- Complex problem solving

Research has shown improvements on several reasoning benchmarks when models are prompted with chain-of-thought demonstrations. 

```python
cot_prompt = """
Solve the following problem.

A student has 5 notebooks.
Each notebook costs $4.
The student also buys 2 pens costing $3 each.

Calculate the total amount spent.

Let's think step by step.
"""

print(cot_prompt)
```

### Expected Reasoning

Notebook cost:

5 × $4 = $20

Pen cost:

2 × $3 = $6

Total:

$20 + $6 = $26

### Final Answer

$26

## 5. System Messages

A system message defines the behavior, role, rules, and constraints of an AI assistant.

It can specify:

- Persona or role
- Tone
- Rules
- Output format
- Restrictions
- Response length

### Example

System Message:

"You are a professional Python tutor.
Explain concepts in simple language.
Always provide a short example.
Use Markdown formatting."

User Message:

"What is a Python list?"

```python
system_message = """
You are a professional Python tutor.

Rules:
1. Explain concepts in simple language.
2. Provide one practical example.
3. Use Markdown.
4. Keep explanations concise.
"""

user_message = "Explain Python lists."

print("SYSTEM MESSAGE:")
print(system_message)

print("\nUSER MESSAGE:")
print(user_message)
```

## 6. Prompt Engineering Pattern

A useful prompt can be structured into several components:

### 1. Persona
Who should the AI act as?

### 2. Instruction
What should the AI do?

### 3. Context
What information does the AI need?

### 4. Format
How should the answer be presented?

### 5. Examples
What does a correct response look like?

### 6. Constraints
What rules or limitations must be followed?

### General Structure

Persona + Instruction + Context + Format + Examples + Constraints

```python
structured_prompt = """
PERSONA:
You are an experienced data science tutor.

INSTRUCTION:
Explain the concept of overfitting.

CONTEXT:
The learner is a beginner in machine learning.

FORMAT:
- Definition
- Simple explanation
- Example
- One prevention technique

CONSTRAINTS:
Use simple English and keep the response under 150 words.
"""

print(structured_prompt)
```

## 7. Complete Prompt Pattern Example

### Format

### Constraint

This structure makes the desired behavior much clearer than simply asking:

"Tell me about overfitting."

## 8. Common Prompt Engineering Pitfalls

### Pitfall 1: Vague Instructions

Bad:
"Tell me about Python."

Better:
"Explain Python lists to a beginner in 5 bullet points with one code example."



### Pitfall 2: Assuming Knowledge

Bad:
"Fix my code."

The model does not have the code.

Better:
Provide the actual code and describe the expected behavior.



### Pitfall 3: No Output Format

Bad:
"Analyze this data."

Better:
"Return the analysis using these sections:
1. Key findings
2. Problems
3. Recommendations"


### Pitfall 4: Conflicting Instructions

Example:

"Give a detailed explanation in exactly 20 words."

The instructions may conflict because "detailed" and "exactly 20 words" can be difficult to satisfy together.


### Pitfall 5: Too Much Unnecessary Context

Providing irrelevant information can make prompts longer and less focused.

Only include context that helps the model complete the task.

```python
bad_prompt = """
Tell me about machine learning.
"""

good_prompt = """
You are a beginner-friendly machine learning instructor.

Explain supervised learning to a university student.

Include:
1. Definition
2. How it works
3. Two examples
4. Difference between classification and regression

Use simple English and bullet points.
"""

print("BAD PROMPT:")
print(bad_prompt)

print("\nGOOD PROMPT:")
print(good_prompt)
```

### Observation

The good prompt is more effective because it clearly specifies:

- Role
- Topic
- Audience
- Required information
- Output format
- Language complexity

Clear instructions reduce ambiguity and make the expected response easier to follow.

```python
prompts = {
    "Zero-Shot": """
Classify this review as Positive, Negative, or Neutral:
"The service was excellent."
""",

    "Few-Shot": """
Classify reviews.

"I love it." -> Positive
"I hate it." -> Negative
"It is okay." -> Neutral

"The service was excellent." ->
""",

    "Chain-of-Thought": """
A shop has 10 apples.
It sells 4 apples and then receives 6 more.

Solve the problem step by step and give the final number.
""",

    "System Prompt": """
SYSTEM:
You are a helpful mathematics tutor.
Explain calculations clearly.

USER:
What is 25% of 200?
"""
}

for technique, prompt in prompts.items():
    print("=" * 60)
    print(technique)
    print(prompt)
```

```python
def call_model(system_message, user_prompt):
    start_time = time.perf_counter()

    response = client.responses.create(
        model="gpt-5.6",
        input=[
            {
                "role": "system",
                "content": system_message
            },
            {
                "role": "user",
                "content": user_prompt
            }
        ]
    )

    elapsed_time = time.perf_counter() - start_time

    return response.output_text.strip(), elapsed_time


print("API function created successfully.")
```

# Task 2: Build Prompt Template Library

The purpose of this task is to create reusable prompt templates for common LLM tasks.

Templates created:

- Summarization
- Information Extraction
- Text Generation
- Text Analysis

Each template contains:

- System message
- User instruction
- Output format
- Constraints
- Examples

Variables used:

- `{text}`
- `{format}`
- `{constraints}`
- `{examples}`

```python
prompt_library = {
    "summarization": {
        "system": (
            "You are an expert summarization assistant. "
            "Create accurate and concise summaries."
        ),
        "user": """
Summarize the following text:

{text}

Output format:
{format}

Constraints:
{constraints}
""",
        "examples": []
    },

    "extraction": {
        "system": (
            "You are an information extraction assistant. "
            "Extract only information explicitly available in the text."
        ),
        "user": """
Extract the requested information from the following text:

{text}

Output format:
{format}

Constraints:
{constraints}

Examples:
{examples}
""",
        "examples": []
    },

    "generation": {
        "system": (
            "You are a professional content generation assistant. "
            "Follow all instructions and constraints."
        ),
        "user": """
Generate content based on the following input:

{text}

Output format:
{format}

Constraints:
{constraints}

Examples:
{examples}
""",
        "examples": []
    },

    "analysis": {
        "system": (
            "You are an analytical assistant. "
            "Provide structured and evidence-based analysis."
        ),
        "user": """
Analyze the following text:

{text}

Output format:
{format}

Constraints:
{constraints}

Examples:
{examples}
""",
        "examples": []
    }
}

print("Prompt library created successfully.")
print("Available templates:", list(prompt_library.keys()))
```

```python
with open("prompt_templates.json", "w", encoding="utf-8") as file:
    json.dump(
        prompt_library,
        file,
        indent=4,
        ensure_ascii=False
    )

print("prompt_templates.json saved successfully.")
```

```python
with open("prompt_templates.json", "r", encoding="utf-8") as file:
    loaded_library = json.load(file)

print(json.dumps(loaded_library, indent=4))
```

```python
def build_prompt(
    task_name,
    text,
    output_format,
    constraints,
    examples=""
):
    template = loaded_library[task_name]

    user_prompt = template["user"].format(
        text=text,
        format=output_format,
        constraints=constraints,
        examples=examples
    )

    return {
        "system": template["system"],
        "user": user_prompt
    }


print("Prompt builder created successfully.")
```

```python
summary_prompt = build_prompt(
    task_name="summarization",
    text=(
        "Artificial intelligence is transforming many industries "
        "by automating repetitive tasks and supporting decision making."
    ),
    output_format="3 bullet points",
    constraints="Use simple English."
)

print("SYSTEM:")
print(summary_prompt["system"])

print("\nUSER:")
print(summary_prompt["user"])
```

```python
extraction_prompt = build_prompt(
    task_name="extraction",
    text=(
        "Areeba is studying Data Science and works on "
        "machine learning projects."
    ),
    output_format='JSON with name, field, and project',
    constraints="Do not infer information that is not provided."
)

print("SYSTEM:")
print(extraction_prompt["system"])

print("\nUSER:")
print(extraction_prompt["user"])
```

```python
generation_prompt = build_prompt(
    task_name="generation",
    text="Write an introduction for a beginner Python course.",
    output_format="One professional paragraph",
    constraints="Use simple and friendly English.",
    examples="Educational and beginner-friendly tone."
)

print("SYSTEM:")
print(generation_prompt["system"])

print("\nUSER:")
print(generation_prompt["user"])
```

```python
analysis_prompt = build_prompt(
    task_name="analysis",
    text=(
        "The company's sales increased by 20% in Q1 "
        "but decreased by 5% in Q2."
    ),
    output_format="Key findings and possible explanations",
    constraints=(
        "Use bullet points and avoid unsupported assumptions."
    )
)

print("SYSTEM:")
print(analysis_prompt["system"])

print("\nUSER:")
print(analysis_prompt["user"])
```

# Task 3: Advanced Output Control

## 3.1 Strict JSON Output

The prompt must enforce:

- Exact schema
- Required fields
- Data types
- No additional fields
- Valid JSON
- Null for missing information

```python
json_prompt = """
You are an information extraction system.

Extract information from the provided text.

Return ONLY valid JSON.

Use exactly this schema:

{{
    "name": "string or null",
    "email": "string or null",
    "skills": ["string"],
    "experience_years": "number or null"
}}

Rules:
- Do not add extra fields.
- Use null when information is missing.
- Do not invent information.
- Do not include Markdown.
- Do not include explanations.
- Return valid JSON only.

Text:
{text}
"""

test_json_input = """
Areeba Amjad is a Data Science student.
She knows Python, Machine Learning and SQL.
"""

print(json_prompt.format(text=test_json_input))
```

```python
markdown_table_prompt = """
You are a data formatting assistant.

Convert the input into a Markdown table.

Use exactly these columns:

| Name | Role | Skill |

Rules:
- Create one row for each person.
- Keep exactly three columns.
- If information is missing, use N/A.
- Keep the Markdown table valid.
- Do not add explanations.
- Do not add extra columns.

Data:
{text}
"""

table_test = """
Areeba is a Data Science student who knows Python.
Ali is a software developer who knows JavaScript.
Sara is a Data Analyst.
"""

print(markdown_table_prompt.format(text=table_test))
```

## 3.3 Controlled Code Generation

The code generation prompt specifies:

- Programming language
- Coding style
- Type hints
- Function name
- Error handling
- Documentation
- Docstrings
- Output restrictions

```python
code_generation_prompt = """
You are a professional Python developer.

Generate Python 3 code for the requested task.

Task:
{text}

Function name:
{function_name}

Expected behavior:
{behavior}

Requirements:
- Follow PEP 8.
- Use meaningful variable names.
- Use type hints.
- Include a Google-style docstring.
- Include basic error handling.
- Create reusable code.
- Do not use unnecessary external libraries.
- Return only Python code.
- Do not include Markdown code fences.
"""

code_test = code_generation_prompt.format(
    text="Calculate the average of a list of numbers.",
    function_name="calculate_average",
    behavior=(
        "Return the average and safely handle an empty list."
    )
)

print(code_test)
```

## 3.4 Robustness Testing

The prompts will be tested against different input conditions:

1. Normal input
2. Empty input
3. Missing information
4. Unexpected format
5. Conflicting information
6. Malformed input
7. Very long input

The objective is to determine whether the prompt continues to follow its required output structure.

```python
def estimate_tokens(text):
    """
    Estimate the number of tokens in English text.

    This is only a rough approximation.
    Actual token counts depend on the tokenizer used by the model.
    """
    return max(1, len(text) // 4)


print("Token estimation function created successfully.")
```
