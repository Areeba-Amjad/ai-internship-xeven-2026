## 1. List vs Tuple vs Set vs Dictionary

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Syntax | `[ ]` | `( )` | `{ }` | `{key: value}` |
| Ordered | Yes | Yes | No guaranteed indexing | Yes (Python 3.7+) |
| Mutable | Yes | No | Yes | Yes |
| Allows duplicates | Yes | Yes | No | Keys: No |
| Indexed | Yes | Yes | No | By key |
| Main use | Collection of changeable items | Fixed collection | Unique items | Key-value data |
| Example | `[1, 2, 3]` | `(1, 2, 3)` | `{1, 2, 3}` | `{"name": "Ali"}` |

## 2. When to Use Each Data Structure

### List
Use a list when:
- You need an ordered collection.
- You need to modify the data.
- Duplicate values are allowed.
- You need indexing and slicing.

### Tuple
Use a tuple when:
- The data should not change.
- You want to represent a fixed collection.
- You need a lightweight immutable structure.

### Set
Use a set when:
- You need unique values.
- You want to remove duplicates.
- You need fast membership testing.
- You need set operations such as union and intersection.

### Dictionary
Use a dictionary when:
- Data is represented using key-value pairs.
- You need fast lookup using a key.
- You want to store structured information.

## 3. Time Complexity

Time complexity describes how the running time of an operation changes as the amount of data increases.

### Common Big-O Notations

- **O(1)** → Constant time
- **O(n)** → Linear time
- **O(log n)** → Logarithmic time

### Average-Case Complexity

| Operation | List | Tuple | Set | Dictionary |
|---|---:|---:|---:|---:|
| Access | O(1) | O(1) | N/A | O(1) by key |
| Search | O(n) | O(n) | O(1) | O(1) by key |
| Insertion | O(1)* | N/A | O(1) | O(1) |
| Deletion | O(n) | N/A | O(1) | O(1) |

`*` List append at the end is O(1) on average.

Note: Some operations can have different worst-case complexities. The table focuses on typical average-case behavior.

## 4. List Example

A list is useful when we need an ordered and mutable collection.

````python
students = ["Ali", "Sara", "Ahmed"]

print("Original list:", students)

students.append("Ayesha")
students[0] = "Hamza"

print("Updated list:", students)
print("First student:", students[0])
````

## 5. Tuple Example

A tuple is useful for storing fixed data that should not be changed.

For example, geographic coordinates can be stored as a tuple.

````python
coordinates = (28.6139, 77.2090)

print("Coordinates:", coordinates)
print("Latitude:", coordinates[0])
print("Longitude:", coordinates[1])
````

## 6. Set Example

A set automatically removes duplicate values and is useful when we only need unique items.

````python
visitors = {"192.168.1.1", "192.168.1.2", "192.168.1.1"}

print("Unique visitors:", visitors)

visitors.add("192.168.1.3")

print("After adding a visitor:", visitors)
print("Is 192.168.1.2 a visitor?", "192.168.1.2" in visitors)
````

## 7. Dictionary Example

A dictionary stores information using key-value pairs.

````python
student = {
    "name": "Ayesha",
    "age": 22,
    "program": "Data Science"
}

print("Student:", student)
print("Name:", student["name"])
print("Program:", student["program"])

student["age"] = 23

print("Updated age:", student["age"])
````

````python
def add_tag(contact_id, tag):
    """Add a tag to a contact."""
    if contact_id not in contacts:
        print("Contact not found.")
        return

    contacts[contact_id]["tags"].add(tag.lower())
    print("Tag added successfully.")


def remove_tag(contact_id, tag):
    """Remove a tag from a contact."""
    if contact_id not in contacts:
        print("Contact not found.")
        return

    contacts[contact_id]["tags"].discard(tag.lower())
    print("Tag removed successfully.")


def find_by_tag(tag):
    """Find contacts containing a specific tag."""
    tag = tag.lower()

    return {
        contact_id: contact
        for contact_id, contact in contacts.items()
        if tag in contact["tags"]
    }
````

````python
def save_to_json(filename="contacts.json"):
    """Save contacts to a JSON file."""
    try:
        data = {}

        for contact_id, contact in contacts.items():
            data[contact_id] = {
                "name": contact["name"],
                "phone": contact["phone"],
                "email": contact["email"],
                "tags": list(contact["tags"]),
                "notes": contact["notes"]
            }

        with open(filename, "w", encoding="utf-8") as file:
            json.dump(data, file, indent=4)

        print("Contacts saved successfully.")

    except OSError as error:
        print(f"Error saving contacts: {error}")


def load_from_json(filename="contacts.json"):
    """Load contacts from a JSON file."""
    global contacts

    try:
        with open(filename, "r", encoding="utf-8") as file:
            data = json.load(file)

        contacts = {}

        for contact_id, contact in data.items():
            contacts[contact_id] = {
                "name": contact["name"],
                "phone": contact["phone"],
                "email": contact["email"],
                "tags": set(contact.get("tags", [])),
                "notes": contact.get("notes", [])
            }

        print("Contacts loaded successfully.")

    except FileNotFoundError:
        print("Contacts file not found.")

    except json.JSONDecodeError:
        print("Invalid JSON file.")

    except (KeyError, TypeError) as error:
        print(f"Invalid contact data: {error}")

    except OSError as error:
        print(f"Error loading contacts: {error}")
````

````python
from collections import Counter


def show_statistics():
    """Display contact statistics."""
    total_contacts = len(contacts)

    tag_counter = Counter()

    for contact in contacts.values():
        tag_counter.update(contact["tags"])

    print("\n--- Contact Statistics ---")
    print(f"Total contacts: {total_contacts}")

    if tag_counter:
        print("\nMost used tags:")

        for tag, count in tag_counter.most_common():
            print(f"{tag}: {count}")

    else:
        print("No tags available.")

    category_counter = Counter()

    for contact in contacts.values():
        for tag in contact["tags"]:
            category_counter[tag] += 1

    print("\nContacts per category:")

    if category_counter:
        for category, count in category_counter.items():
            print(f"{category}: {count}")
    else:
        print("No categories available.")
````

````python
def display_contacts(contact_data):
    """Display contacts in a readable format."""
    if not contact_data:
        print("No contacts found.")
        return

    for contact_id, contact in contact_data.items():
        print("\n-------------------------")
        print(f"ID: {contact_id}")
        print(f"Name: {contact['name']}")
        print(f"Phone: {contact['phone']}")
        print(f"Email: {contact['email']}")
        print(f"Tags: {', '.join(sorted(contact['tags']))}")
        print(f"Notes: {', '.join(contact['notes'])}")
````
