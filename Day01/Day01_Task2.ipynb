## 7. Types of Artificial Intelligence

Artificial Intelligence can be broadly discussed in terms of Narrow
AI and General AI.

### 7.1 Narrow AI

Narrow AI, also called Weak AI, is designed to perform specific tasks.
It operates within a limited area and does not possess general human
intelligence.

Most AI systems used in the real world today are examples of Narrow
AI.

**Examples:**

- Recommendation systems on Netflix and YouTube
- Voice assistants such as Siri and Alexa
- Fraud detection systems in banking
- AI-based medical image analysis
- Chatbots and customer service systems

### 7.2 General AI

Artificial General Intelligence (AGI) is a future conceptual form of
AI that would be capable of understanding, learning, and performing a
wide range of intellectual tasks across different domains, similar
to human intelligence.

AGI does not currently exist as a demonstrated general-purpose
technology.

### Narrow AI vs. General AI

| Feature | Narrow AI | General AI |
|---|---|---|
| Scope | Specific tasks | Broad range of tasks |
| Current Status | Widely used today | Future conceptual goal |
| Learning | Limited to its designed tasks | Intended to learn across domains |
| Example | Recommendation system | Human-level general intelligence |

### Key Insight

The main difference between Narrow AI and General AI is their scope.
Narrow AI is designed for specific tasks and is widely used today,
while General AI refers to a future concept involving broader,
human-like intellectual capabilities.

## 9. Fields of Artificial Intelligence

Artificial Intelligence includes several specialized fields that
focus on different aspects of intelligent systems.

### 9.1 Machine Learning

Machine Learning (ML) is a field of AI in which computer systems learn
patterns from data and use those patterns to make predictions or
decisions without being explicitly programmed for every situation.

**Examples:**

- Fraud detection
- House price prediction
- Recommendation systems
- Customer classification

### 9.2 Deep Learning

Deep Learning is a specialized area of Machine Learning that uses
multi-layered neural networks to learn complex patterns from large
amounts of data.

Deep Learning is commonly used for tasks involving images, speech,
text, and other complex data.

**Examples:**

- Image recognition
- Speech recognition
- Natural Language Processing
- Autonomous driving systems

### 9.3 AI Engineering

AI Engineering focuses on building, integrating, deploying, and
maintaining AI systems in real-world environments.

An AI Engineer works on the complete process of turning AI models
into reliable and usable applications.

**Common AI Engineer Responsibilities:**

- Collecting and preparing data
- Selecting and developing suitable AI models
- Training and evaluating models
- Integrating AI models into applications
- Deploying AI systems
- Monitoring model performance
- Maintaining and improving AI systems
- Troubleshooting AI-related issues

### Key Insight

Machine Learning allows systems to learn from data, while Deep
Learning uses neural networks to learn complex patterns. AI
Engineering focuses on applying these technologies to build,
deploy, monitor, and maintain practical AI systems.

## 11. Role of an AI Engineer in 2026

An AI Engineer is responsible for developing and implementing
Artificial Intelligence systems that solve real-world problems.

In 2026, the role of an AI Engineer involves not only developing AI
models but also deploying, monitoring, maintaining, and improving AI
systems in production environments.

### Key Responsibilities of an AI Engineer

- **Building AI Systems:** Develop and integrate AI models into
  software applications.
- **Data Preparation:** Collect, clean, transform, and prepare data
  for AI model development.
- **Model Development:** Select appropriate algorithms and train
  machine learning or deep learning models.
- **Model Evaluation:** Test models using appropriate performance
  metrics and validate their results.
- **Deployment:** Deploy trained AI models so they can be used by
  real users and applications.
- **Monitoring:** Monitor deployed AI systems to identify performance
  problems, errors, or changes in data.
- **Maintenance:** Update and improve AI systems as new data,
  requirements, or technologies become available.
- **Responsible AI:** Consider reliability, privacy, security,
  fairness, and responsible use when developing AI systems.

### AI Engineer Workflow

The typical workflow can be summarized as:

**Data → Development → Training → Evaluation → Deployment →
Monitoring → Maintenance**

### Key Insight

The role of an AI Engineer extends beyond training a model. An AI
Engineer is involved in the complete AI lifecycle, from preparing
data and developing models to deploying, monitoring, maintaining, and
improving AI systems.

## 12. Training vs. Inference

Training and inference are two important stages in the lifecycle of
an AI or Machine Learning model.

### 12.1 Training

Training is the process in which an AI model learns patterns from
historical or training data.

During training, the model analyzes input data and adjusts its
parameters to improve its performance.

**Example:**

A fraud detection model can be trained using historical transactions
that are labeled as legitimate or fraudulent. The model learns
patterns associated with fraudulent transactions.

### 12.2 Inference

Inference is the process of using a trained AI model to make
predictions or decisions on new, unseen data.

The model does not learn from the new input during normal inference.
Instead, it uses the patterns learned during training.

**Example:**

When a new financial transaction occurs, a trained fraud detection
model can analyze it and predict whether the transaction is likely
to be fraudulent.

### Training vs. Inference

| Feature | Training | Inference |
|---|---|---|
| Purpose | Learn patterns from data | Make predictions |
| Data | Training data | New or unseen data |
| Model | Parameters are updated | Parameters are normally fixed |
| Process | Usually more computationally intensive | Usually optimized for fast predictions |
| Example | Learning fraud patterns | Detecting fraud in a new transaction |

### Key Insight

Training teaches an AI model by learning patterns from data, while
inference uses the trained model to make predictions on new data.

In simple terms:

**Training = Learning**

**Inference = Prediction**
