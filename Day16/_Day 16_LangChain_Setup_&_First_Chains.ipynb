# 1. What is LangChain?

LangChain is an open-source framework designed to help developers build applications powered by Large Language Models (LLMs).

Instead of interacting with an LLM through a single API call, LangChain provides tools and abstractions for connecting different components such as models, prompts, documents, retrievers, tools, and application workflows.

LangChain can simplify the development of applications such as:

- AI chatbots
- Question-answering systems
- Retrieval-Augmented Generation (RAG) applications
- Document analysis systems
- Summarization applications
- AI agents
- Automated LLM workflows

### Why is LangChain useful?

A raw LLM API can be sufficient for a simple question-and-answer application. However, as an application becomes more complex, developers may need to connect multiple steps and external data sources.

LangChain provides reusable components that make these workflows easier to organize and maintain.

### Important Point

LangChain is **not an LLM itself**.

It is a framework that helps developers build applications around LLMs.

# 3. Core Components of LangChain

The major concepts/components of LangChain include:

1. Models
2. Prompts
3. Chains

5. Document Loaders
6. Retrievers
7. Tools

These components can be combined to create complete LLM-powered applications.

## 3.1 Models

Models are the components responsible for generating or processing language.

LangChain can work with models from different providers.

Examples include:

- OpenAI models
- Anthropic Claude models
- Google Gemini models
- Hugging Face models
- Other supported LLM providers

### Role of a Model

The model receives an input and produces an output.

For example:

Input:
"Explain machine learning in simple words."

Output:
"Machine learning is a method where computers learn patterns from data..."

LangChain provides interfaces that make it easier to integrate different models into an application.

## 3.2 Prompts

A prompt is the instruction or input given to a language model.

LangChain provides prompt templates that allow developers to create reusable prompts.

### Example

Instead of writing:

"Explain Python in simple words."

We can create a template:

"Explain {topic} in simple words."

Then we can provide different values:

- topic = Python
- topic = Machine Learning
- topic = LangChain

This makes prompts reusable and easier to manage.

## 3.3 Chains

A chain represents a sequence of operations that are executed together.

For example:

Input
↓
Prompt
↓
LLM
↓
Output Parser
↓
Final Answer

Chains are useful when an application requires multiple predictable steps.

For example, a document summarization application might:

1. Load a document.
2. Split the document into smaller sections.
3. Send sections to an LLM.
4. Generate summaries.
5. Combine the summaries.

LangChain makes it easier to organize these steps into reusable workflows.

## 3.4 Memory / Conversation History

LLMs do not automatically remember every previous interaction in an application.

For a chatbot, previous messages can be stored and included in later requests so that the application can maintain conversational context.

For example:

User:
"My name is Areeba."

Assistant:
"Nice to meet you, Areeba!"

User:
"What is my name?"

If the application maintains conversation history, it can use the previous message to answer:

"Your name is Areeba."

Modern LangChain applications commonly manage conversation history using message-history and runnable-based components.

# 4. LangChain Expression Language (LCEL)

LangChain Expression Language, commonly called **LCEL**, is a way to compose LangChain components into workflows.

LCEL allows developers to connect runnable components using a simple pipe operator:

`|`

### Example

```text
Prompt | Model | Output Parser


## 4.1 The Pipe `|` Operator

The pipe operator is one of the most important features of LCEL.

It connects two or more runnable components.

Example:

```python
chain = prompt | model | parser


## 4.2 Runnable Interface

LCEL components follow a common Runnable interface.

A Runnable represents something that can receive an input and produce an output.

Examples of runnable components include:

- Prompt templates
- Models
- Output parsers
- Lambda functions
- Retrievers
- Other processing steps

Because these components follow a common interface, they can be easily combined into chains.

# 5. Document Loaders

Document loaders are used to import information from external sources into a LangChain application.

They convert information into a standardized document representation that can be processed by other LangChain components.

### Examples of data sources

- PDF files
- Text files
- Websites
- CSV files
- Word documents
- Databases
- Cloud storage
- Other supported data sources

### Typical Workflow

Document Source
↓
Document Loader
↓
Document Objects
↓
Text Splitting
↓
Embeddings / Processing
↓
Retrieval or LLM Application

## 5.1 Why are Document Loaders Important?

LLMs normally cannot automatically access every document or private dataset available to a user.

Document loaders help applications bring external information into an LLM workflow.

For example, suppose we have a university PDF containing course information.

A LangChain application can:

1. Load the PDF.
2. Extract its text.
3. Split the text into smaller chunks.
4. Store or process the chunks.
5. Retrieve relevant information.
6. Ask an LLM to generate an answer.

This approach is commonly used in Retrieval-Augmented Generation (RAG) systems.

# 7. When Should We Use LangChain?

LangChain is especially useful for applications that involve complex LLM workflows.

### Good use cases

#### 1. RAG Applications
Applications that answer questions using information from external documents.

#### 2. Document Processing
Applications that need to read and analyze PDFs, websites, files, or databases.

#### 3. Multi-Step Workflows
Applications where multiple processing steps must be connected.

#### 4. AI Agents
Applications where an LLM needs to decide which tools or actions to use.

#### 5. Structured Applications
Applications that need reusable prompts, parsers, retrievers, tools, and other components.

### When LangChain may not be necessary

For a simple application such as:

```text
User Question → One API Call → Answer

## Generic Document Loader

````python
from pathlib import Path

from langchain_community.document_loaders import (
    TextLoader,
    PyPDFLoader,
    CSVLoader,
)
````

````python
def load_document(file_path):
    """Load a document using the appropriate LangChain loader."""
    
    extension = Path(file_path).suffix.lower()

    if extension == ".txt":
        loader = TextLoader(file_path, encoding="utf-8")
    elif extension == ".pdf":
        loader = PyPDFLoader(file_path)
    elif extension == ".csv":
        loader = CSVLoader(file_path=file_path)
    else:
        raise ValueError(
            f"Unsupported file type: {extension}"
        )

    return loader.load()
````

````python
text_docs = load_document("sample_document.txt")

print("Text documents loaded:", len(text_docs))
print(text_docs[0].page_content[:300])
````

## Generic Document Loader

A generic `load_document()` function was created to automatically
select the appropriate LangChain document loader based on the
file extension.

Supported formats:

- `.txt` → TextLoader
- `.pdf` → PyPDFLoader
- `.csv` → CSVLoader

This approach makes the document-loading workflow reusable and
reduces the need to manually create a different loader for every file type.

# Task 3: Build Document Q&A Chain

The goal of this task is to build a simple document question-answering
workflow using LangChain.

Workflow:

````python
document = load_document("sample_document.txt")

document_text = "\n".join(
    doc.page_content for doc in document
)

print("Document loaded successfully!")
print(document_text)
````

````python
from langchain_core.prompts import PromptTemplate

print("PromptTemplate imported successfully!")
````

````python
qa_prompt = PromptTemplate(
    input_variables=["context", "question"],
    template="""
You are a helpful assistant.

Use the following document to answer the question.

Document:
{context}

Question:
{question}

Answer:
"""
)

print("Q&A prompt created successfully!")
````

````python
from langchain_core.runnables import RunnableLambda


def local_model(prompt_text):
    """Simulate an LLM response locally."""
    return (
        "LangChain is a framework for building applications "
        "using Large Language Models (LLMs)."
    )


model = RunnableLambda(local_model)

print("Local model created successfully!")
````

````python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

print("Output parser created successfully!")
````

````python
qa_chain = qa_prompt | model | output_parser

print("Q&A chain created successfully!")
````

````python
document = load_document("sample_document.txt")

document_text = "\n".join(
    doc.page_content for doc in document
)

print("Document loaded successfully!")
print(document_text)
````

````python
question = "What is LangChain?"

answer = qa_chain.invoke({
    "context": document_text,
    "question": question
})

print("Question:", question)
print("Answer:", answer)
````

## Handling Long Documents

Large documents may exceed the context window of an LLM.

For this basic implementation, the document can be truncated before
being passed to the model.

In production applications, documents are commonly split into smaller
chunks and relevant chunks are retrieved before sending them to the LLM.

````python
def prepare_document(text, max_characters=4000):
    """Limit document size before sending it to the model."""
    
    if len(text) > max_characters:
        print("Warning: Document was truncated.")
        return text[:max_characters]
    
    return text


short_document = prepare_document(document_text)

print("Document characters:", len(short_document))
````

````python
question = "What are the key components of LangChain?"

answer = qa_chain.invoke({
    "context": short_document,
    "question": question
})

print("Question:", question)
print("Answer:", answer)
````
