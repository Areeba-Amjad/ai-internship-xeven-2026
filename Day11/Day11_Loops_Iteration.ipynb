# 2. While Loops

A `while` loop is used to repeatedly execute a block of code as long as a given condition remains `True`.

Unlike a `for` loop, which is commonly used to iterate over a sequence, a `while` loop is mainly condition-based. The loop continues until its condition becomes `False`.

###  Source 1: ChatGPT

#### While Loops in Python

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

## 4. Nested Loops

A nested loop is a loop placed inside another loop.

The outer loop executes first, and for every iteration of the outer loop, the inner loop executes completely.

Nested loops are commonly used for:

- Matrix operations
- 2D lists
- Tables
- Grids
- 2D data processing

When two loops each depend on the size of the input, nested iteration can result in approximately O(n²) time complexity.

# Task 1: Data Processing Pipeline

### Objectives

- Process a list of 1000 records using a for loop.
- Apply transformations to valid records.
- Use `enumerate()` to track progress.
- Use `continue` to skip invalid records.
- Use `break` to stop processing after a critical error.
- Use `zip()` to combine multiple data sources.
- Validate that multiple data sources have matching lengths.

````python
records = []

for i in range(1, 1001):
    records.append({
        "id": i,
        "name": f"User_{i}",
        "score": (i * 7) % 101
    })

print("Total records:", len(records))
print("First record:", records[0])
````

````python
processed_records = []

for index, record in enumerate(records, start=1):
    print(f"Record #{index} of {len(records)} processed")

    if record["score"] < 0:
        continue

    transformed_record = {
        "id": record["id"],
        "name": record["name"].upper(),
        "score": record["score"] / 100
    }

    processed_records.append(transformed_record)

print("\nTotal processed:", len(processed_records))
````

# Task 2: Pattern Generators

### Objectives

- Generate multiplication tables using nested loops.
- Generate pyramid patterns.
- Generate number triangles.
- Perform matrix operations.
- Transpose a matrix.
- Calculate row and column sums.
- Find diagonal elements.
- Create ASCII art using nested loops and conditionals.

````python
number = 5

print(f"Multiplication Table of {number}")

for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
````

````python
for table in range(2, 5):
    print(f"\nTable of {table}")

    for multiplier in range(1, 6):
        print(f"{table} x {multiplier} = {table * multiplier}")
````

````python
rows = 5

for i in range(1, rows + 1):
    spaces = " " * (rows - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
````

````python
rows = 5

for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(j, end=" ")

    print()
````

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Original Matrix:")

for row in matrix:
    print(row)
````

````python
rows = len(matrix)
columns = len(matrix[0])

transpose = []

for column in range(columns):
    new_row = []

    for row in range(rows):
        new_row.append(matrix[row][column])

    transpose.append(new_row)

print("Transposed Matrix:")

for row in transpose:
    print(row)
````

````python
row_sums = []

for row in matrix:
    total = 0

    for value in row:
        total += value

    row_sums.append(total)

print("Row sums:", row_sums)
````

````python
column_sums = []

for column in range(len(matrix[0])):
    total = 0

    for row in range(len(matrix)):
        total += matrix[row][column]

    column_sums.append(total)

print("Column sums:", column_sums)
````

````python
main_diagonal = []

for i in range(len(matrix)):
    main_diagonal.append(matrix[i][i])

print("Main diagonal:", main_diagonal)
````

````python
size = 7

for row in range(size):
    for column in range(size):

        if row == 0 or row == size - 1:
            print("*", end=" ")

        elif column == 0 or column == size - 1:
            print("*", end=" ")

        else:
            print(" ", end=" ")

    print()
````

# Task 3: Number Analysis System

### Objectives

- Find prime numbers up to N using a while loop.
- Use mathematical optimization for prime checking.
- Calculate factorial using an iterative approach.
- Generate Fibonacci numbers iteratively.
- Build a number guessing game.
- Limit the number of guessing attempts.
- Provide hints to the user.

````python
def is_prime(number):
    if number < 2:
        return False

    divisor = 2

    while divisor * divisor <= number:
        if number % divisor == 0:
            return False

        divisor += 1

    return True
````

````python
n = 50
number = 2
primes = []

while number <= n:
    if is_prime(number):
        primes.append(number)

    number += 1

print("Prime numbers up to", n)
print(primes)
````

````python
number = 5
factorial = 1
counter = 1

while counter <= number:
    factorial *= counter
    counter += 1

print(f"Factorial of {number} = {factorial}")
````

````python
terms = 10

first = 0
second = 1
counter = 0

fibonacci = []

while counter < terms:
    fibonacci.append(first)

    next_value = first + second
    first = second
    second = next_value

    counter += 1

print("Fibonacci sequence:")
print(fibonacci)
````

### Number Guessing Game

The game generates a secret number and gives the player a limited number of attempts.

The player receives hints:

- "Too high" if the guess is greater than the secret number.
- "Too low" if the guess is smaller than the secret number.
- "Correct" if the player guesses the number successfully.

````python
import random

secret_number = random.randint(1, 100)
attempts = 7
won = False

print("Guess the number between 1 and 100.")
print(f"You have {attempts} attempts.")

for attempt in range(1, attempts + 1):
    guess = int(input(f"Attempt {attempt}: Enter your guess: "))

    if guess == secret_number:
        print("Congratulations! You guessed correctly.")
        won = True
        break

    if guess < secret_number:
        print("Hint: Too low.")

    else:
        print("Hint: Too high.")

if not won:
    print(f"Game over! The correct number was {secret_number}.")
    
````
