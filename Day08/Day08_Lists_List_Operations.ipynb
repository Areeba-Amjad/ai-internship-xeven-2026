# 3. List Slicing

## 3.1 ChatGPT Research

List slicing is used to access a portion of a list.

The general syntax is:

`list[start:end:step]`

The start index is included, while the end index is normally excluded.
Python also supports negative indexing, which allows elements to be accessed
from the end of a list.

## 3.2 Google Gemini Research

Gemini explains that slicing allows programmers to extract a sublist without
changing the original list.

Examples include:

- `numbers[:3]` — first three elements
- `numbers[2:5]` — elements from index 2 to 4
- `numbers[-3:]` — last three elements
- `numbers[::2]` — every second element

## 3.3 Claude Research

Claude explains that slicing is particularly useful when working with
sequences and when only a selected portion of a list is required.

The optional step value determines how many positions Python moves between
selected elements.

# Task 1: Student Grade Manager

````python
# Student Grade Manager

students = [
    {"name": "Ali", "grade": 85},
    {"name": "Sara", "grade": 92},
    {"name": "Ahmed", "grade": 78},
    {"name": "Ayesha", "grade": 95},
    {"name": "Hassan", "grade": 88}
]

print("Initial Student Records:")
for student in students:
    print(student)
````

````python
# Add a new student

students.append({"name": "Zain", "grade": 90})

print("After Adding Zain:")
for student in students:
    print(student)
````

````python
# Remove a student

students = [student for student in students if student["name"] != "Ahmed"]

print("After Removing Ahmed:")
for student in students:
    print(student)
````

````python
# Update Sara's grade

for student in students:
    if student["name"] == "Sara":
        student["grade"] = 96

print("After Updating Sara's Grade:")
for student in students:
    print(student)
````

````python
# Calculate average grade

grades = [student["grade"] for student in students]

average_grade = sum(grades) / len(grades)

print("Average Grade:", round(average_grade, 2))
````

````python
# Find highest and lowest grades

highest_grade = max(students, key=lambda student: student["grade"])
lowest_grade = min(students, key=lambda student: student["grade"])

print("Highest Grade:", highest_grade)
print("Lowest Grade:", lowest_grade)
````

````python
# Sort students by grade

sorted_students = sorted(
    students,
    key=lambda student: student["grade"],
    reverse=True
)

print("Students Sorted by Grade:")
for student in sorted_students:
    print(student)
````

````python
# Display top 3 performers

top_3 = sorted_students[:3]

print("Top 3 Performers:")
for student in top_3:
    print(student)
````

````python
# Filter students with grade 90 or above

high_performers = [
    student for student in students
    if student["grade"] >= 90
]

print("Students with Grade 90 or Above:")
for student in high_performers:
    print(student)
````

### Task 1 Conclusion

The Student Grade Manager demonstrated how Python lists can be used to store and manage student records. List methods, list comprehensions, sorting, filtering, and slicing were applied to add, remove, update, and analyze student grades.

# Task 2: Shopping cart data

````python
# Shopping cart data

item_names = []
item_prices = []
item_quantities = []

print("Shopping cart created successfully.")
````

````python
def add_item(name, price, quantity):
    item_names.append(name)
    item_prices.append(price)
    item_quantities.append(quantity)
    print(f"{name} added to cart.")


# Add sample items
add_item("Laptop Bag", 35, 1)
add_item("Wireless Mouse", 25, 2)
add_item("Keyboard", 30, 1)
add_item("USB Cable", 10, 3)
add_item("Notebook", 5, 2)
````

````python
print("Current Shopping Cart:")

for i in range(len(item_names)):
    print(
        f"{item_names[i]} - "
        f"Price: ${item_prices[i]} - "
        f"Quantity: {item_quantities[i]}"
    )
````

````python
def remove_item(name):
    if name in item_names:
        index = item_names.index(name)

        item_names.pop(index)
        item_prices.pop(index)
        item_quantities.pop(index)

        print(f"{name} removed from cart.")
    else:
        print(f"{name} not found in cart.")


# Remove an item
remove_item("USB Cable")
````

````python
def update_quantity(name, new_quantity):
    if name in item_names:
        index = item_names.index(name)
        item_quantities[index] = new_quantity

        print(
            f"{name} quantity updated to {new_quantity}."
        )
    else:
        print(f"{name} not found in cart.")


# Update quantity
update_quantity("Wireless Mouse", 3)
````

````python
def calculate_total():
    total = 0

    for i in range(len(item_names)):
        subtotal = item_prices[i] * item_quantities[i]
        total += subtotal

    return total


total = calculate_total()

print(f"Cart Total: ${total:.2f}")
````

````python
total = calculate_total()

if total > 100:
    discount = total * 0.10
    final_total = total - discount
else:
    discount = 0
    final_total = total

print(f"Subtotal: ${total:.2f}")
print(f"Discount: ${discount:.2f}")
print(f"Final Total: ${final_total:.2f}")
````

````python
print("=" * 50)
print("              SHOPPING RECEIPT")
print("=" * 50)

for i in range(len(item_names)):
    subtotal = item_prices[i] * item_quantities[i]

    print(
        f"{item_names[i]:20} "
        f"Qty: {item_quantities[i]:2} "
        f"Price: ${item_prices[i]:6.2f} "
        f"Subtotal: ${subtotal:7.2f}"
    )

print("-" * 50)
print(f"Subtotal:        ${total:.2f}")
print(f"Discount:        ${discount:.2f}")
print(f"Final Total:     ${final_total:.2f}")
print("=" * 50)
````

````python
recent_items = item_names[-3:]

print("Recently Added Items:")
for item in recent_items:
    print(item)
````

## Task 3: Data Cleaning Pipeline

````python
# Create messy data

messy_data = [
    "  Apple  ",
    "banana",
    "APPLE",
    None,
    " Orange ",
    "banana",
    "  mango",
    None,
    "MANGO ",
    " grape"
]

print("Original Messy Data:")
print(messy_data)
````

````python
# Before cleaning metrics

total_before = len(messy_data)
unique_before = len(set(messy_data))
non_null_before = sum(
    1 for item in messy_data if item is not None
)
completeness_before = (
    non_null_before / total_before
) * 100

print("Before Cleaning:")
print("Total items:", total_before)
print("Unique items:", unique_before)
print("Non-null items:", non_null_before)
print(f"Completeness: {completeness_before:.2f}%")
````

````python
# Remove None values using list comprehension

cleaned_data = [
    item for item in messy_data
    if item is not None
]

print("After Removing None Values:")
print(cleaned_data)
````

````python
# Remove extra spaces using list comprehension

cleaned_data = [
    item.strip()
    for item in cleaned_data
]

print("After Removing Extra Spaces:")
print(cleaned_data)
````

````python
# Normalize case using list comprehension

cleaned_data = [
    item.lower()
    for item in cleaned_data
]

print("After Normalizing Case:")
print(cleaned_data)
````

````python
# Remove duplicates using list comprehension

cleaned_data = [
    item
    for index, item in enumerate(cleaned_data)
    if item not in cleaned_data[:index]
]

print("After Removing Duplicates:")
print(cleaned_data)
````

````python
# After cleaning metrics

total_after = len(cleaned_data)
unique_after = len(set(cleaned_data))
non_null_after = sum(
    1 for item in cleaned_data if item is not None
)

completeness_after = (
    non_null_after / total_after
) * 100

print("After Cleaning:")
print("Total items:", total_after)
print("Unique items:", unique_after)
print("Non-null items:", non_null_after)
print(f"Completeness: {completeness_after:.2f}%")
````

````python
# Before and after comparison

print("=" * 45)
print("       DATA QUALITY COMPARISON")
print("=" * 45)

print(f"{'Metric':<20} {'Before':<12} {'After':<12}")
print("-" * 45)

print(
    f"{'Total items':<20} "
    f"{total_before:<12} "
    f"{total_after:<12}"
)

print(
    f"{'Unique items':<20} "
    f"{unique_before:<12} "
    f"{unique_after:<12}"
)

print(
    f"{'Non-null items':<20} "
    f"{non_null_before:<12} "
    f"{non_null_after:<12}"
)

print(
    f"{'Completeness':<20} "
    f"{completeness_before:.2f}%{'':<7} "
    f"{completeness_after:.2f}%"
)

print("=" * 45)
````

````python
# Display final cleaned data

print("Original Data:")
print(messy_data)

print("\nFinal Cleaned Data:")
print(cleaned_data)
````
