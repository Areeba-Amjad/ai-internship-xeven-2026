# Dictionary Methods

Python provides several built-in methods for working with dictionaries:

- get()
- keys()
- values()
- items()
- update()
- pop()

```python
student = {
    "name": "Areeba",
    "age": 22,
    "field": "Data Science"
}

print("get():", student.get("name"))
print("keys():", student.keys())
print("values():", student.values())
print("items():", student.items())

student.update({"city": "Rahim Yar Khan"})
print("After update():", student)

student.pop("age")
print("After pop():", student)
```

# Nested Dictionaries

A nested dictionary is a dictionary that contains another dictionary as
a value. It is useful for representing complex and structured data.

```python
students = {
    "student1": {
        "name": "Areeba",
        "age": 22,
        "field": "Data Science"
    },
    "student2": {
        "name": "Sara",
        "age": 21,
        "field": "Computer Science"
    }
}

print(students)
print(students["student1"]["name"])
```

# JSON Format

JSON stands for JavaScript Object Notation.

JSON is commonly used for storing and exchanging structured data.

Python provides the `json` module for working with JSON files.

Important functions:

- json.load() — reads JSON data from a file
- json.dump() — writes Python data to a JSON file

```python
import json

student = {
    "name": "Areeba",
    "age": 22,
    "field": "Data Science"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

print("JSON file created successfully.")
```

```python
import json

with open("student.json", "r") as file:
    data = json.load(file)

print(data)
print(data["name"])
```

# Dictionary Comprehensions

Dictionary comprehension provides a concise way to create dictionaries.

### Syntax

{k: v for k, v in pairs}

```python
numbers = [1, 2, 3, 4, 5]

squares = {number: number ** 2 for number in numbers}

print(squares)
```

## Task 1: Student Information System

### Objective

Create a Student Information System using nested dictionaries.

The system will:
- Store student information using student IDs.
- Store grades for different subjects.
- Add new students.
- Update student grades.
- Calculate student averages.
- Find the top-performing student.
- Save and load student data using JSON.
- Generate a performance report sorted by GPA.

```python
students = {
    "S001": {
        "name": "Areeba",
        "age": 22,
        "grades": {
            "Python": 90,
            "Statistics": 85,
            "Machine Learning": 88
        }
    },
    "S002": {
        "name": "Sara",
        "age": 21,
        "grades": {
            "Python": 82,
            "Statistics": 89,
            "Machine Learning": 91
        }
    },
    "S003": {
        "name": "Ali",
        "age": 23,
        "grades": {
            "Python": 78,
            "Statistics": 80,
            "Machine Learning": 84
        }
    }
}

print(students)
```

### Function: add_student()

This function adds a new student to the nested dictionary using a unique
student ID.

```python
def add_student(student_id, name, age, grades):
    students[student_id] = {
        "name": name,
        "age": age,
        "grades": grades
    }
    print(f"Student {name} added successfully.")


add_student(
    "S004",
    "Hamza",
    22,
    {
        "Python": 87,
        "Statistics": 84,
        "Machine Learning": 90
    }
)

print(students["S004"])
```

### Function: update_grade()

This function updates the grade of a specific subject for a student.

```python
def update_grade(student_id, subject, score):
    if student_id in students:
        students[student_id]["grades"][subject] = score
        print("Grade updated successfully.")
    else:
        print("Student not found.")


update_grade("S001", "Python", 95)

print(students["S001"]["grades"])
```

### Function: get_student_average()

This function calculates the average grade of a student.

```python
def get_student_average(student_id):
    if student_id not in students:
        return None

    grades = students[student_id]["grades"].values()

    if not grades:
        return 0

    return sum(grades) / len(grades)


average = get_student_average("S001")

print(f"S001 Average: {average:.2f}")
```

### Function: find_top_student()

This function compares student averages and returns the student with the
highest average score.

```python
def find_top_student():
    if not students:
        return None

    top_student = max(
        students,
        key=get_student_average
    )

    return top_student


top_id = find_top_student()

print("Top Student ID:", top_id)
print("Top Student Name:", students[top_id]["name"])
print(
    "Top Student Average:",
    f"{get_student_average(top_id):.2f}"
)
```

### Save Student Data to JSON

The `json.dump()` function is used to save the student dictionary into a
JSON file. This provides data persistence.

```python
import json

with open("students.json", "w") as file:
    json.dump(students, file, indent=4)

print("Student data saved successfully.")
```

### Load Student Data from JSON

The `json.load()` function reads the saved JSON data and converts it back
into a Python dictionary.

```python
with open("students.json", "r") as file:
    loaded_students = json.load(file)

students = loaded_students

print("Student data loaded successfully.")
print(students)
```

### Student Performance Report

The report calculates the GPA/average for every student and sorts students
from highest to lowest performance.

```python
report = []

for student_id, student in students.items():
    average = get_student_average(student_id)

    report.append({
        "student_id": student_id,
        "name": student["name"],
        "gpa": round(average, 2)
    })

report.sort(key=lambda student: student["gpa"], reverse=True)

for student in report:
    print(
        f"{student['student_id']} - "
        f"{student['name']} - "
        f"GPA: {student['gpa']}"
    )
```

# Task 2: Product Inventory Manager

### Objective

Create a product inventory system using dictionaries.

The system will:
- Store product information.
- Add new products.
- Update stock.
- Search products by category.
- Generate low-stock alerts.
- Calculate total inventory value.
- Calculate average product price per category.
- Export inventory data as JSON.

```python
inventory = {
    "P001": {
        "name": "Laptop",
        "price": 120000,
        "quantity": 5,
        "category": "Electronics"
    },
    "P002": {
        "name": "Mouse",
        "price": 2500,
        "quantity": 15,
        "category": "Electronics"
    },
    "P003": {
        "name": "Notebook",
        "price": 500,
        "quantity": 30,
        "category": "Stationery"
    },
    "P004": {
        "name": "Pen",
        "price": 100,
        "quantity": 50,
        "category": "Stationery"
    }
}

print(inventory)
```

```python
def add_product(product_id, name, price, quantity, category):
    inventory[product_id] = {
        "name": name,
        "price": price,
        "quantity": quantity,
        "category": category
    }

    print(f"{name} added successfully.")


add_product(
    "P005",
    "Keyboard",
    4500,
    10,
    "Electronics"
)

print(inventory["P005"])
```

```python
def update_stock(product_id, quantity):
    if product_id in inventory:
        inventory[product_id]["quantity"] = quantity
        print("Stock updated successfully.")
    else:
        print("Product not found.")


update_stock("P002", 8)

print(inventory["P002"])
```

```python
def search_by_category(category):
    results = {}

    for product_id, product in inventory.items():
        if product["category"].lower() == category.lower():
            results[product_id] = product

    return results


electronics = search_by_category("Electronics")

print(electronics)
```

```python
def low_stock_alert(threshold=10):
    low_stock = {}

    for product_id, product in inventory.items():
        if product["quantity"] <= threshold:
            low_stock[product_id] = product

    return low_stock


alerts = low_stock_alert()

print("Low Stock Products:")

for product_id, product in alerts.items():
    print(
        f"{product_id} - {product['name']} - "
        f"Quantity: {product['quantity']}"
    )
```

### Calculate Total Inventory Value

Inventory value is calculated using:

`price × quantity`

```python
total_value = sum(
    product["price"] * product["quantity"]
    for product in inventory.values()
)

print(f"Total Inventory Value: Rs. {total_value:,.2f}")
```

```python
from collections import defaultdict

category_prices = defaultdict(list)

for product in inventory.values():
    category_prices[product["category"]].append(
        product["price"]
    )

print("Average Product Price by Category:")

for category, prices in category_prices.items():
    average_price = sum(prices) / len(prices)

    print(
        f"{category}: Rs. {average_price:,.2f}"
    )
```

### Export Inventory Report as JSON

```python
with open("inventory.json", "w") as file:
    json.dump(inventory, file, indent=4)

print("Inventory report exported successfully.")
```

# Task 3: Configuration Manager

### Objective

Create a configuration manager that stores application settings in a
JSON file.

The system will:
- Store database settings.
- Store API keys.
- Store feature settings.
- Load configuration.
- Validate required keys.
- Provide default values.
- Update configuration programmatically.
- Handle missing keys using `.get()`.

```python
config = {
    "database": {
        "host": "localhost",
        "port": 5432,
        "name": "student_db"
    },
    "api_keys": {
        "weather_api": "demo_key"
    },
    "features": {
        "dark_mode": True,
        "notifications": True
    }
}

with open("config.json", "w") as file:
    json.dump(config, file, indent=4)

print("config.json created successfully.")
```

```python
def load_config(filename="config.json"):
    try:
        with open(filename, "r") as file:
            return json.load(file)
    except FileNotFoundError:
        print("Configuration file not found.")
        return {}


config = load_config()

print(config)
```

```python
def validate_required_keys(config, required_keys):
    missing_keys = []

    for key in required_keys:
        if key not in config:
            missing_keys.append(key)

    return missing_keys


required_keys = [
    "database",
    "api_keys",
    "features"
]

missing = validate_required_keys(
    config,
    required_keys
)

if missing:
    print("Missing keys:", missing)
else:
    print("All required keys are available.")
```

### Handling Missing Keys with `.get()`

The `.get()` method allows us to safely access a dictionary key. If the key
does not exist, a default value can be returned instead of causing an error.

```python
database = config.get("database", {})

host = database.get("host", "localhost")
port = database.get("port", 5432)
database_name = database.get(
    "name",
    "default_database"
)

print("Host:", host)
print("Port:", port)
print("Database:", database_name)
```

```python
def update_config(config, section, key, value):
    if section not in config:
        config[section] = {}

    config[section][key] = value

    print(
        f"{section}.{key} updated successfully."
    )


update_config(
    config,
    "features",
    "dark_mode",
    False
)

print(config["features"])
```

```python
def save_config(config, filename="config.json"):
    with open(filename, "w") as file:
        json.dump(config, file, indent=4)

    print("Configuration saved successfully.")


save_config(config)
```
